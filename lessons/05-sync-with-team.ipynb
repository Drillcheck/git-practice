# 05 팀의 최신 변경과 맞추기

나는 주문 합계에 봉사료 10%를 더하는 커밋을 만들었습니다.
같은 시간에 팀원 Emily는 합계를 센트 단위로 반올림하는 커밋을 만들었습니다.
두 사람 모두 `kitchen/order.py`의 같은 줄을 고쳤습니다.

이제 두 변경을 하나로 모아야 합니다. 방법은 두 가지입니다.

- **rebase**: 내 커밋을 Emily의 커밋 위로 옮겨서 이력을 한 줄로 만듭니다.
- **merge**: 두 갈래를 그대로 두고, 합치는 커밋을 하나 새로 만듭니다.

이 장에서는 두 방법을 다 해 봅니다. 같은 줄을 고쳤으니 어느 쪽이든 충돌이 납니다. 충돌을 푸는 것까지가 이 장의 실습입니다.
마지막에는 미션 m04를 merge로 풀고 Pull Request로 채점받습니다.

## 1. 원본 저장소도 등록하기

내 컴퓨터의 저장소는 내 fork를 clone한 것입니다. 그래서 지금은 내 fork 하나만 알고 있습니다.
그런데 팀의 최신 변경은 원본 저장소(`Drillcheck/git-practice`)에 먼저 올라옵니다. 원본에서도 받아 올 수 있게 등록해 둡니다.

### 1-1 등록된 원격 저장소 보기

입력:

```
git remote -v
```

결과:

```
origin	https://github.com/<내 아이디>/git-practice.git (fetch)
origin	https://github.com/<내 아이디>/git-practice.git (push)
```

`origin`은 clone할 때 git이 붙여 준 이름입니다. 내 fork를 가리킵니다.

### 1-2 원본을 upstream으로 등록

원본 저장소에는 보통 `upstream`이라는 이름을 붙입니다. 이름은 마음대로 지어도 되지만, 대부분의 팀이 이 이름을 씁니다.

입력:

```
git remote add upstream https://github.com/Drillcheck/git-practice.git
git remote -v
```

결과:

```
origin	https://github.com/<내 아이디>/git-practice.git (fetch)
origin	https://github.com/<내 아이디>/git-practice.git (push)
upstream	https://github.com/Drillcheck/git-practice.git (fetch)
upstream	https://github.com/Drillcheck/git-practice.git (push)
```

| 이름 | 가리키는 곳 | 하는 일 |
|---|---|---|
| `origin` | 내 fork | 내 작업을 push하는 곳 |
| `upstream` | 원본 저장소 | 팀의 최신 변경을 받아 오는 곳 |

`upstream`에는 push하지 않습니다. 원본 저장소에 쓸 권한이 없고, 변경은 Pull Request로만 보냅니다.

### 1-3 원본의 브랜치 받아 오기

입력:

```
git fetch upstream
```

결과:

```
From https://github.com/Drillcheck/git-practice
 * [new branch]      main         -> upstream/main
 * [new branch]      mission/m01  -> upstream/mission/m01
 * [new branch]      mission/m02  -> upstream/mission/m02
 * [new branch]      mission/m03  -> upstream/mission/m03
 * [new branch]      mission/m04  -> upstream/mission/m04
 * [new branch]      teammate/m04 -> upstream/teammate/m04
```

`fetch`는 원격 저장소의 커밋을 받아서 `upstream/main` 같은 이름으로 저장해 둡니다. **내 브랜치와 내 파일은 하나도 바뀌지 않습니다.** 받아 온 것을 보고 나서, 내 브랜치에 합칠지 언제 합칠지는 내가 정합니다.

### 1-4 받아 온 브랜치 보기

입력:

```
git branch -r
```

결과:

```
  origin/HEAD -> origin/main
  origin/main
  origin/mission/m01
  origin/mission/m02
  origin/mission/m03
  origin/mission/m04
  origin/teammate/m04
  upstream/HEAD -> upstream/main
  upstream/main
  upstream/mission/m01
  upstream/mission/m02
  upstream/mission/m03
  upstream/mission/m04
  upstream/teammate/m04
```

`-r`은 원격(remote) 브랜치만 보여 줍니다. `origin/`으로 시작하는 것은 내 fork에서, `upstream/`으로 시작하는 것은 원본에서 받아 온 것입니다.

`upstream/HEAD -> upstream/main`은 브랜치가 아니라 "원본의 기본 브랜치는 `main`"이라는 표시입니다. 처음 `fetch`할 때 git이 적어 둡니다.

### 1-5 fetch와 pull

`git pull`은 `fetch`와 합치기를 한 번에 합니다.

| 명령 | 하는 일 |
|---|---|
| `git fetch` | 받아 오기만 합니다. 내 브랜치는 그대로입니다 |
| `git pull` | 받아 온 뒤 지금 브랜치에 merge합니다 |
| `git pull --rebase` | 받아 온 뒤 지금 브랜치를 그 위로 rebase합니다 |

처음 보는 변경이 올라왔을 때는 `fetch`로 받아서 먼저 읽어 보는 게 안전합니다. 이 장에서는 `fetch`를 쓰고, `pull --rebase`는 4절에서 한 번 씁니다.

## 3. rebase: 내 커밋을 팀원의 커밋 위로 옮기기

rebase는 내 커밋을 떼어 내서 다른 커밋 위에 다시 붙입니다. 결과는 처음부터 Emily의 커밋 다음에 작업한 것처럼 한 줄로 이어진 이력입니다.

미션 브랜치는 5절에서 merge로 풀어야 합니다. rebase는 연습용 브랜치를 새로 만들어서 해 봅니다.

### 3-1 연습용 브랜치 만들기

입력:

```
git switch -c m04-rebase
```

결과:

```
Switched to a new branch 'm04-rebase'
```

### 3-2 Emily의 커밋 위로 rebase

입력:

```
git rebase origin/teammate/m04
```

결과:

```
Rebasing (1/1)
Auto-merging kitchen/order.py
CONFLICT (content): Merge conflict in kitchen/order.py
error: could not apply 189ef3b... Add a 10% service charge
hint: Resolve all conflicts manually, mark them as resolved with
hint: "git add/rm <conflicted_files>", then run "git rebase --continue".
hint: You can instead skip this commit: run "git rebase --skip".
hint: To abort and get back to the state before "git rebase", run "git rebase --abort".
hint: Disable this message with "git config set advice.mergeConflict false"
Could not apply 189ef3b... # Add a 10% service charge
```

`CONFLICT`가 보이면 충돌입니다. 같은 줄을 두 사람이 다르게 고쳐서, git이 어느 쪽을 남길지 정하지 못하고 멈췄습니다. 고장 난 것이 아니라, 사람에게 판단을 넘긴 것입니다.

### 3-3 지금 상태 보기

입력:

```
git status
```

결과:

```
interactive rebase in progress; onto 6cbe174
Last command done (1 command done):
   pick 189ef3b # Add a 10% service charge
No commands remaining.
You are currently rebasing branch 'm04-rebase' on '6cbe174'.
  (fix conflicts and then run "git rebase --continue")
  (use "git rebase --skip" to skip this patch)
  (use "git rebase --abort" to check out the original branch)

Changes to be committed:
  (use "git restore --staged <file>..." to unstage)
	modified:   kitchen/test_order.py

Unmerged paths:
  (use "git restore --staged <file>..." to unstage)
  (use "git add <file>..." to mark resolution)
	both modified:   kitchen/order.py
```

| 부분 | 뜻 |
|---|---|
| `interactive rebase in progress` | rebase가 중간에 멈춰 있습니다 |
| `Changes to be committed` | 충돌 없이 옮겨진 변경. `test_order.py`는 Emily가 건드리지 않아서 그대로 옮겨졌습니다 |
| `both modified` | 충돌이 난 파일. 이 파일을 내가 고쳐야 합니다 |

멈춘 상태에서 빠져나오고 싶으면 `git rebase --abort`를 입력합니다. rebase 전으로 완전히 돌아갑니다.

### 3-4 충돌 난 파일 열어 보기

편집기로 `kitchen/order.py`를 엽니다. 터미널에서 내용만 보려면 이렇게 입력합니다.

입력:

```
cat kitchen/order.py
```

결과:

```
from kitchen.menu import MENU


def total(items):
<<<<<<< HEAD
    return round(sum(MENU[name] * qty for name, qty in items.items()), 2)
=======
    subtotal = sum(MENU[name] * qty for name, qty in items.items())
    return subtotal * 1.1  # service charge 10%
>>>>>>> 189ef3b (Add a 10% service charge)
```

git이 두 사람의 줄을 모두 남기고 표시를 붙여 두었습니다.

| 표시 | 뜻 |
|---|---|
| `<<<<<<< HEAD` 부터 `=======` 까지 | 지금 올라타는 쪽. rebase에서는 **Emily의 커밋**입니다 |
| `=======` 부터 `>>>>>>> 189ef3b` 까지 | 옮겨 붙이는 쪽. rebase에서는 **내 커밋**입니다 |

> rebase에서 `HEAD`는 내 쪽이 아닙니다. 내 커밋을 Emily의 커밋 위에 하나씩 다시 쌓는 중이라, 지금 서 있는 곳(HEAD)이 Emily의 커밋입니다. 5절의 merge에서는 반대가 됩니다.

### 3-5 두 변경을 모두 살려서 고치기

한쪽을 고르면 다른 사람의 변경이 사라집니다. 봉사료를 더한 합계를 반올림하면 두 사람의 의도가 다 남습니다.
편집기에서 표시 세 줄을 지우고, `total` 함수를 아래처럼 고친 뒤 저장합니다.

```python
from kitchen.menu import MENU


def total(items):
    subtotal = sum(MENU[name] * qty for name, qty in items.items())
    return round(subtotal * 1.1, 2)  # service charge 10%, rounded to cents
```


저장한 뒤 `<<<<<<<`, `=======`, `>>>>>>>`가 파일에 하나도 남지 않았는지 다시 확인합니다.

### 3-6 rebase 이어 가기

고친 파일을 `git add`로 표시하면 "이 파일의 충돌은 풀었다"는 뜻이 됩니다. 그다음 rebase를 이어 갑니다.

입력:

```
git add kitchen/order.py
git rebase --continue
```

결과:

```
[detached HEAD ???????] Add a 10% service charge
 Author: You <you@example.invalid>
 2 files changed, 3 insertions(+), 2 deletions(-)
Successfully rebased and updated refs/heads/m04-rebase.
```

`Author: You`라는 줄이 보입니다. 미션의 시작 커밋은 "나"라는 역할로 미리 만들어 둔 것이라서 작성자가 `You`로 되어 있습니다. rebase는 작성자를 그대로 두고, 커밋을 다시 만든 사람만 내 이름으로 남깁니다.

## 6. merge와 rebase, 언제 무엇을

| 상황 | 쓰는 것 | 이유 |
|---|---|---|
| 아직 나 혼자 쓰는 브랜치를 최신 main에 맞출 때 | rebase | 이력이 한 줄이라 리뷰어가 읽기 쉽습니다 |
| 다른 사람도 받아 간 브랜치 (`main`, 같이 쓰는 기능 브랜치) | merge | 기존 커밋을 바꾸지 않아서 남의 이력을 깨뜨리지 않습니다 |
| 팀원의 작업 기록을 갈래째 남겨야 할 때 | merge | 언제 갈라져서 언제 합쳤는지가 이력에 남습니다 |
| rebase나 amend 뒤에 push할 때 | `git push --force-with-lease` | 그사이 누가 push했으면 거절해 줍니다 |

기준은 하나입니다. **다른 사람이 이미 받아 간 커밋은 다시 쓰지 않습니다.** rebase와 amend는 커밋을 새로 만들어 바꿔 끼우기 때문에, 그 커밋을 받아 간 사람의 이력과 어긋납니다.
팀마다 정해 둔 방식이 있으면 그 방식을 따릅니다. PR을 merge할 때 GitHub에서 고르는 방식(merge, squash, rebase)도 팀이 정합니다.

### 6-1 연습용 브랜치 정리하기

결과:

```
To https://github.com/<내 아이디>/git-practice.git
 - [deleted]         m04-rebase
Deleted branch m04-rebase (was ???????).
```

`-D`는 "merge되지 않은 커밋이 있어도 지운다"는 뜻입니다. 연습용이라 지워도 되는 커밋입니다.
퀴즈 05-5에서 `m04-ours`를 만들었다면 `git branch -D m04-ours`로 함께 지웁니다.

## 7. 내 fork의 main을 원본에 맞추기

fork는 만든 순간의 복사본입니다. 원본 `main`에 새 커밋이 올라와도 내 fork의 `main`은 그대로 있습니다. 새 작업을 시작하기 전에 내 `main`을 원본에 맞춰 두면, 새 브랜치가 최신 코드에서 출발합니다.

### 7-1 뒤처진 상황 만들기

원본에 새 커밋이 올라오기를 기다릴 수는 없으니, 거꾸로 내 `main`을 한 커밋 뒤로 돌려서 같은 상태를 만듭니다. 그다음 fork에도 그대로 올립니다.

입력:

```
git switch main
git reset --hard HEAD~1
git push --force-with-lease origin main
```

결과:

```
Switched to branch 'main'
Your branch is up to date with 'origin/main'.
HEAD is now at 885bc94 Add the grader
To https://github.com/<내 아이디>/git-practice.git
 + ???????...885bc94 main -> main (forced update)
```

이제 내 컴퓨터의 `main`과 fork의 `main`은 원본보다 커밋 하나가 뒤에 있습니다. fork를 오래전에 만들어 둔 사람의 상태와 같습니다.

### 7-2 원본과 비교하기

입력:

```
git fetch upstream
git log --oneline --graph main upstream/main -3
```

결과:

```
* ??????? Add the practice guide
* 885bc94 Add the grader
* 1bddc8d Add a note about the weekend dinner set
```

`upstream/main`이 맨 위에 있고, 그 바로 아래가 내 `main`입니다. 내 `main`에는 원본에 없는 커밋이 하나도 없습니다. 원본이 앞으로 더 나가 있을 뿐입니다.

> 그사이 원본에 새 커밋이 올라왔다면 `upstream/main` 쪽에 커밋이 더 보입니다. 이어지는 단계는 똑같습니다.

### 7-3 fast-forward로 맞추기

입력:

```
git merge --ff-only upstream/main
```

결과:

```
Updating 885bc94..???????
Fast-forward
```

그 아래로 바뀐 파일 목록이 이어집니다. 원본의 `main` 마지막 커밋에 들어 있는 파일들입니다.

`Fast-forward`는 **새 커밋을 만들지 않고 브랜치 이름만 앞으로 옮겼다**는 뜻입니다. 내 `main`에 원본에 없는 커밋이 없으니, 원본의 커밋을 그대로 따라가면 됩니다.
`--ff-only`는 "이렇게 이름만 옮길 수 있을 때만 하고, 아니면 아무것도 하지 말라"는 조건입니다. 7-5에서 이 조건이 거절하는 경우를 봅니다.

### 7-4 fork에도 올리기

입력:

```
git push origin main
git status
```

결과:

```
To https://github.com/<내 아이디>/git-practice.git
   885bc94..???????  main -> main
On branch main
Your branch is up to date with 'origin/main'.

nothing to commit, working tree clean
```

이번에는 `--force-with-lease` 없이 올라갑니다. fork의 `main`에 있던 커밋을 지우지 않고 뒤에 붙이기만 하기 때문입니다. `up to date with 'origin/main'`이면 내 컴퓨터, fork, 원본의 `main`이 모두 같습니다.

내 기능 브랜치가 있다면, 이렇게 맞춘 `main` 위로 3절처럼 `git rebase main`을 해서 옮깁니다. 7-7에서 한 번 해 봅니다.

> GitHub에서 내 fork 페이지를 열면 **Sync fork** 버튼으로도 fork의 `main`을 원본에 맞출 수 있습니다. GitHub CLI로는 `gh repo sync <내 아이디>/git-practice -b main`입니다. 어느 쪽이든 GitHub 위의 fork만 바뀝니다. 내 컴퓨터의 `main`은 그다음 `git pull`로 받아야 합니다.

```
- 다음 주에 메뉴 사진 넣기
```


입력:

```
git add NOTES.md
git commit -m "Add my notes"
git push --force-with-lease origin main
```

결과:

```
[main ???????] Add my notes
 1 file changed, 1 insertion(+)
 create mode 100644 NOTES.md
To https://github.com/<내 아이디>/git-practice.git
 + ???????...??????? main -> main (forced update)
```

이제 원본으로 맞춰 봅니다.

그래프가 두 갈래입니다. 내 `main`에는 원본에 없는 `Add my notes`가 있고, 원본에는 내 `main`에 없는 커밋이 있습니다. 이름만 앞으로 옮겨서는 둘 다 살릴 수 없으니 `--ff-only`가 거절했습니다(`Not possible to fast-forward`). hint는 merge(`git merge --no-ff`)와 rebase(`git rebase`) 중 하나를 고르라고 알려 줍니다.

### 7-6 git pull은 어떻게 하나

같은 상황에서 `git pull`을 해 봅니다.

입력:

```
git pull upstream main
```

결과:

```
From https://github.com/Drillcheck/git-practice
 * branch            main       -> FETCH_HEAD
hint: You have divergent branches and need to specify how to reconcile them.
hint: You can do so by running one of the following commands sometime before
hint: your next pull:
hint:
hint:   git config pull.rebase false  # merge
hint:   git config pull.rebase true   # rebase
hint:   git config pull.ff only       # fast-forward only
hint:
hint: You can replace "git config" with "git config --global" to set a default
hint: preference for all repositories. You can also pass --rebase, --no-rebase,
hint: or --ff-only on the command line to override the configured default per
hint: invocation.
fatal: Need to specify how to reconcile divergent branches.
```

git 2.27부터 `git pull`은 두 갈래가 갈라져 있으면 멈추고, 어떻게 합칠지 정하라고 합니다. 고를 수 있는 것은 세 가지입니다.

| 설정 | 갈라졌을 때 `git pull`이 하는 일 |
|---|---|
| `git config --global pull.rebase false` | merge 커밋을 만들어 합칩니다 |
| `git config --global pull.rebase true` | 내 커밋을 받아 온 커밋 위로 rebase합니다 |
| `git config --global pull.ff only` | 아무것도 하지 않고 멈춥니다. fast-forward일 때만 받습니다 |

이 과정에서는 `pull.ff only`를 씁니다. `git pull`이 몰래 merge 커밋을 만들거나 커밋을 다시 쓰는 일이 없고, 갈라졌을 때는 6절의 표를 보고 merge와 rebase 중 하나를 내가 직접 고르게 됩니다. 4-8처럼 rebase하고 싶을 때는 `git pull --rebase`라고 그때그때 적습니다.

입력:

```
git config --global pull.ff only
git pull upstream main
```

결과:

```
From https://github.com/Drillcheck/git-practice
 * branch            main       -> FETCH_HEAD
hint: Diverging branches can't be fast-forwarded, you need to either:
hint:
hint: 	git merge --no-ff
hint:
hint: or:
hint:
hint: 	git rebase
hint:
hint: Disable this message with "git config set advice.diverging false"
fatal: Not possible to fast-forward, aborting.
```

이번에는 설정을 고르라는 안내 대신, 7-5의 `git merge --ff-only`와 같은 메시지로 멈춥니다. 무엇으로 합칠지는 내가 정합니다.

### 7-7 메모는 브랜치로 옮기고 main은 원본에 맞추기

`main`은 원본을 따라가는 자리로만 씁니다. 내 작업은 브랜치에 둡니다. 그래야 `main`을 언제든 원본에 맞출 수 있고, PR도 브랜치 단위로 열 수 있습니다.

먼저 메모 커밋이 사라지지 않게 지금 위치에 브랜치 이름을 붙입니다. 그다음 `main`을 원본과 똑같이 맞춥니다.

입력:

```
git branch my-notes
git reset --hard upstream/main
git push --force-with-lease origin main
```

결과:

```
HEAD is now at ??????? Add the practice guide
To https://github.com/<내 아이디>/git-practice.git
 + ???????...??????? main -> main (forced update)
```

`reset --hard upstream/main`은 내 `main`을 원본의 `main`과 똑같이 만들고, `main`에만 있던 커밋은 `main`에서 뺍니다. 그래서 바로 앞에서 `my-notes`로 커밋을 붙잡아 두었습니다. fork의 `main`에 있던 `Add my notes`도 바꿔 끼워야 하므로 `--force-with-lease`로 올렸습니다.
이 방법은 **내 fork의 `main`처럼 내 작업을 두지 않기로 한 브랜치**에만 씁니다. 같이 쓰는 브랜치에서 `reset --hard`하고 force push하면 남의 커밋이 사라집니다.

> `--force-with-lease`에 `--force-if-includes`를 함께 붙이면 한 가지를 더 확인합니다. fork에 있는 커밋을 내가 정말 받아서 내 브랜치에 반영했는지입니다. `git fetch`만 하고 확인하지 않은 커밋을 덮어쓰는 실수를 막아 줍니다.

이제 메모 브랜치를 맞춘 `main` 위로 옮깁니다.

입력:

```
git switch my-notes
git rebase main
git log --oneline --graph -3
```

결과:

```
Switched to branch 'my-notes'
Rebasing (1/1)
Successfully rebased and updated refs/heads/my-notes.
* ??????? Add my notes
* ??????? Add the practice guide
* 885bc94 Add the grader
```

메모 커밋이 원본의 최신 커밋 바로 위에 올라갔습니다. 이 브랜치를 push하면 그대로 PR을 열 수 있는 모양입니다.

연습이 끝났으니 메모 브랜치를 지우고 미션 브랜치로 돌아옵니다.

입력:

```
git switch mission/m04
git branch -D my-notes
```

결과:

```
Switched to branch 'mission/m04'
Your branch is up to date with 'origin/mission/m04'.
Deleted branch my-notes (was ???????).
```

### 퀴즈 05-7

7-1에서는 `--force-with-lease`가 필요했고, 7-4에서는 그냥 `git push`로 올라갔습니다. 두 push에서 fork의 `main`에는 각각 무슨 일이 생겼습니까?

## 8. 충돌에 원래 줄도 함께 보기

충돌 표시는 기본으로 두 쪽만 보여 줍니다. 두 사람이 바꾸기 **전**의 줄도 함께 보면 각자 무엇을 바꿨는지 더 잘 보입니다. git 2.35부터 쓸 수 있는 `zdiff3` 표시 방식입니다.

### 8-1 표시 방식 바꾸기

입력:

```
git config --global merge.conflictStyle zdiff3
```

아무것도 나오지 않으면 성공입니다.

> `zdiff3`는 git 2.35부터 있습니다. 00장 2-1에서 2.44 이상을 설치했다면 문제없습니다. 더 낮은 git에서 이 설정을 켜면 이후 `git switch`, `git commit`, `git merge` 같은 명령이 `fatal: unknown style 'zdiff3' given for 'merge.conflictstyle'`로 멈춥니다. 그때는 `git config --global --unset merge.conflictStyle`로 설정을 지웁니다.

### 8-2 같은 충돌을 다시 내 보기

5절의 충돌을 연습용 브랜치에서 한 번 더 냅니다.

입력:

```
git switch -c m04-zdiff3 upstream/mission/m04
git merge origin/teammate/m04
```

결과:

```
Switched to a new branch 'm04-zdiff3'
branch 'm04-zdiff3' set up to track 'upstream/mission/m04'.
Auto-merging kitchen/order.py
CONFLICT (content): Merge conflict in kitchen/order.py
Automatic merge failed; fix conflicts and then commit the result.
```

입력:

```
cat kitchen/order.py
```

결과:

```
from kitchen.menu import MENU


def total(items):
<<<<<<< HEAD
    subtotal = sum(MENU[name] * qty for name, qty in items.items())
    return subtotal * 1.1  # service charge 10%
||||||| 962df7e
    return sum(MENU[name] * qty for name, qty in items.items())
=======
    return round(sum(MENU[name] * qty for name, qty in items.items()), 2)
>>>>>>> origin/teammate/m04
```

5-3과 비교해 보세요. `|||||||`와 `=======` 사이가 새로 생겼습니다. 두 사람이 갈라지기 전, 원래 있던 줄입니다. `|||||||` 옆의 `962df7e`가 두 갈래가 갈라진 커밋입니다. 원래 줄과 한쪽씩 비교하면 내가 바꾼 것과 Emily가 바꾼 것이 따로 보입니다.

### 8-3 정리하기

이번에는 풀지 않고 merge를 취소한 뒤 연습용 브랜치를 지웁니다.

입력:

```
git merge --abort
git switch mission/m04
git branch -D m04-zdiff3
```

결과:

```
Switched to branch 'mission/m04'
Your branch is up to date with 'origin/mission/m04'.
Deleted branch m04-zdiff3 (was 189ef3b).
```

`merge.conflictStyle` 설정은 그대로 둡니다. 앞으로 생기는 충돌은 모두 이 모양으로 나옵니다.

### 퀴즈 05-8

8-2의 `|||||||` 아래 줄과 위아래 두 쪽을 비교해서, 나와 Emily가 각각 무엇을 바꿨는지 한 줄씩 적어 보세요.

## 정리

| 하고 싶은 것 | 명령 |
|---|---|
| 원본 저장소 등록 | `git remote add upstream 주소` |
| 원격 저장소 목록 | `git remote -v` |
| 받아 오기만 | `git fetch upstream` |
| 원격 브랜치 목록 | `git branch -r` |
| 원격이 여러 개일 때 브랜치 만들어 이동 | `git switch --track origin/브랜치` |
| 갈라진 브랜치 함께 보기 | `git log --oneline --graph 브랜치1 브랜치2` |
| 내 커밋을 다른 브랜치 위로 옮기기 | `git rebase 브랜치` |
| 충돌을 푼 뒤 rebase 이어 가기 | `git add 파일`, `git rebase --continue` |
| rebase 취소 | `git rebase --abort` |
| 두 갈래 합치기 | `git merge 브랜치` |
| 충돌을 푼 뒤 merge 마치기 | `git add 파일`, `git commit` |
| 다시 쓴 이력 올리기 | `git push --force-with-lease` |
| fork의 main을 원본에 맞추기 | `git fetch upstream`, `git switch main`, `git merge --ff-only upstream/main`, `git push origin main` |
| main에 잘못 한 커밋을 버리고 원본과 똑같이 | `git reset --hard upstream/main`, `git push --force-with-lease origin main` |
| `git pull`이 갈라졌을 때 멈추게 | `git config --global pull.ff only` |
| 충돌에 원래 줄도 보기 | `git config --global merge.conflictStyle zdiff3` |
| 받아 와서 그 위로 맞추기 | `git pull --rebase` |
| 원격 브랜치 지우기 | `git push origin --delete 브랜치` |
