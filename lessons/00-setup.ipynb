# 00 준비

이 장을 마치면 세 가지가 된 상태입니다.

- 내 컴퓨터의 터미널이 내 GitHub 계정으로 로그인되어 있습니다.
- 내 컴퓨터에 이 저장소가 있고, 내 fork에 push할 권한이 있습니다.
- kitchen 프로젝트의 테스트가 통과합니다.

## 따라 하는 방법

- 회색 상자 **입력**의 명령을 터미널에 그대로 입력하고 Enter를 누릅니다. 한 번에 한 줄씩 입력합니다.
- `"내 이름"`처럼 따옴표 안의 한국어는 내 정보로 바꿔서 입력합니다. `<내 아이디>`는 `<`와 `>`까지 지우고 내 GitHub 아이디를 넣습니다.
- 상자 **결과**는 예시입니다. 버전 숫자, 아이디, 경로는 내 컴퓨터에 따라 다르게 나옵니다.

## 1. 터미널 열기

| 운영체제 | 여는 방법 |
|---|---|
| Windows | git을 설치한 뒤(2절) 시작 메뉴에서 **Git Bash**를 엽니다. 이 수업의 명령은 모두 Git Bash에서 입력합니다. PowerShell이나 명령 프롬프트가 아닙니다. |
| macOS | `Cmd + Space`를 누르고 "터미널"을 입력해 엽니다. |
| Linux | 배포판의 터미널 앱을 엽니다. |

## 2. git 설치 확인

### 2-1 git 버전 보기

입력:

```
git --version
```

결과 (예시):

```
git version 2.43.0
```

이 수업은 **2.44 이상**이 필요합니다. 그보다 낮은 버전에서는 04장과 05장의 명령 몇 개가 오류 없이 다르게 동작합니다. 숫자가 2.44 이상이면 2-2로 넘어갑니다.

2.44보다 낮으면 새 버전을 설치합니다. 설치한 뒤에는 터미널을 닫았다가 다시 열고 2-1을 다시 입력합니다.

- **Windows**: [git-scm.com/download/win](https://git-scm.com/download/win)에서 최신 Git for Windows를 받아 설치합니다. 기존 설치 위에 그대로 설치하면 됩니다.
- **macOS**: `brew install git`. Homebrew가 없으면 [brew.sh](https://brew.sh)의 설치 명령을 먼저 입력합니다.
- **Linux (Ubuntu)**: `sudo add-apt-repository ppa:git-core/ppa`, `sudo apt update`, `sudo apt install git`을 차례로 입력합니다.

`command not found`가 나오면 설치합니다.

- **Windows**: [git-scm.com/download/win](https://git-scm.com/download/win)에서 받아 설치합니다. 설치 화면의 선택지는 모두 기본값 그대로 두고 Next를 누르면 됩니다. 설치가 끝나면 Git Bash를 열고 2-1을 다시 입력합니다.
- **macOS**: 2-1을 입력했을 때 뜨는 설치 창에서 **설치**를 누릅니다. 창이 뜨지 않으면 `xcode-select --install`을 입력합니다.
- **Linux (Ubuntu, Debian)**: `sudo apt install git`

### 2-2 Python 버전 보기

kitchen 프로젝트의 테스트를 돌리려면 Python 3이 필요합니다.

입력:

```
python3 --version
```

결과 (예시):

```
Python 3.12.3
```

Windows에서 이 명령이 안 되면 `python --version`을 입력해 보세요. 이 경우 이 수업의 `python3`은 모두 `python`으로 바꿔 입력합니다.
둘 다 안 되면 [python.org](https://www.python.org/downloads/)에서 설치합니다. Windows 설치 첫 화면의 **Add python.exe to PATH**를 꼭 체크합니다.

## 3. GitHub 로그인

로그인은 두 곳에서 합니다.

- **웹 브라우저**: fork를 만들고 PR을 열 때 씁니다.
- **터미널**: 내 컴퓨터에서 GitHub로 push할 때 씁니다. git은 GitHub 비밀번호를 받지 않으므로, GitHub CLI(`gh`)로 로그인해 둡니다.

### 3-1 웹에서 로그인

[github.com](https://github.com)에 로그인합니다. 계정이 없다면 **Sign up**으로 만들고, 가입 메일의 인증까지 마칩니다.

### 3-2 GitHub CLI 설치 확인

입력:

```
gh --version
```

결과 (예시):

```
gh version 2.62.0 (2024-11-14)
https://github.com/cli/cli/releases/tag/v2.62.0
```

`command not found`가 나오면 설치합니다.

- **Windows**: [cli.github.com](https://cli.github.com)에서 설치 파일을 받아 설치합니다. 설치가 끝나면 Git Bash를 **닫았다가 다시 열고** 3-2를 다시 입력합니다.
- **macOS**: Homebrew가 있으면 `brew install gh`, 없으면 [cli.github.com](https://cli.github.com)에서 설치 파일을 받습니다.
- **Linux (Ubuntu, Debian)**: `sudo apt install gh`. 패키지를 찾지 못하면 [cli.github.com](https://cli.github.com)의 Linux 안내를 따릅니다.

### 3-3 지금 로그인되어 있는지 확인

입력:

```
gh auth status
```

로그인한 적이 없으면 이렇게 나옵니다. gh 버전에 따라 둘째 문장은 조금 다를 수 있습니다. 3-4로 넘어갑니다.

```
You are not logged into any GitHub hosts. To log in, run: gh auth login
```

이미 로그인되어 있으면 `Logged in to github.com account <내 아이디>`가 나옵니다. 아이디가 내 것이 맞으면 3-4는 건너뜁니다. 다른 사람 아이디라면 `gh auth logout`을 입력한 뒤 3-4를 합니다.

### 3-4 터미널에서 로그인

입력:

```
gh auth login
```

질문이 하나씩 나옵니다. 화살표 키로 고르고 Enter를 누릅니다.

| 질문 | 고를 것 |
|---|---|
| Where do you use GitHub? | **GitHub.com** |
| What is your preferred protocol for Git operations on this host? | **HTTPS** |
| Authenticate Git with your GitHub credentials? | **Yes** |
| How would you like to authenticate GitHub CLI? | **Login with a web browser** |

그러면 8자리 코드가 나옵니다.

```
! First copy your one-time code: ABCD-1234
Press Enter to open https://github.com/login/device in your browser...
```

1. 코드를 적어 두고 Enter를 누릅니다. 브라우저가 열립니다. 열리지 않으면 주소를 직접 엽니다.
2. 브라우저에 코드를 입력하고 **Authorize github**를 누릅니다.
3. 터미널로 돌아오면 이렇게 끝나 있습니다.

```
✓ Authentication complete.
- gh config set -H github.com git_protocol https
✓ Configured git protocol
✓ Logged in as <내 아이디>
```

> 학교나 회사의 공용 컴퓨터라면 수업이 끝날 때 `gh auth logout`으로 로그아웃하세요. 로그인해 두면 이 컴퓨터를 쓰는 누구나 내 계정으로 push할 수 있습니다.

### 3-5 로그인 다시 확인

입력:

```
gh auth status
```

결과 (예시):

```
github.com
  ✓ Logged in to github.com account <내 아이디> (keyring)
  - Active account: true
  - Git operations protocol: https
  - Token: gho_************************************
```

`Logged in to github.com account` 뒤에 내 아이디가 보이면 됩니다.

## 4. git에 내 이름 등록

커밋마다 작성자 이름과 이메일이 함께 저장됩니다.

> **이 이름과 이메일은 로그인이 아닙니다.** git은 여기에 무엇을 쓰든 확인하지 않습니다. 커밋에 "누가 썼는지" 적어 두는 표시일 뿐입니다. GitHub에 올릴 권한은 3절의 로그인이 정합니다.

### 4-1 지금 설정 확인

등록하기 전에 이미 들어 있는 값이 있는지 봅니다. 다른 사람이 쓰던 컴퓨터라면 그 사람 이름이 들어 있을 수 있습니다.

입력:

```
git config --global --list
```

결과 (예시, 3-4에서 로그인한 직후):

```
credential.https://github.com.helper=
credential.https://github.com.helper=!/usr/bin/gh auth git-credential
```

- `credential` 줄은 3-4의 로그인이 넣은 것입니다. "GitHub에 push할 때 gh의 로그인을 쓴다"는 뜻입니다. 지우지 않습니다.
- `user.name`과 `user.email` 줄이 없으면 아직 등록하지 않은 것입니다. 4-2로 갑니다.
- 이미 있는데 내 정보가 맞으면 4-4로 건너뜁니다. 다른 사람 정보라면 4-2와 4-3으로 덮어씁니다.
- `fatal: unable to read config file`이 나오면 설정 파일이 아직 없는 것입니다. 오류가 아닙니다. 4-2로 갑니다.

### 4-2 이름 등록

입력:

```
git config --global user.name "내 이름"
```

아무것도 출력되지 않으면 성공입니다. git은 성공했을 때 말이 없는 명령이 많습니다.

### 4-3 이메일 등록

> **커밋에 넣은 이메일은 공개됩니다.** GitHub에 올린 커밋의 이메일은 누구나 볼 수 있습니다.
> 개인 이메일을 드러내고 싶지 않으면 GitHub 웹에서 **Settings → Emails**로 갑니다. **Keep my email addresses private**를 켜면 `숫자+아이디@users.noreply.github.com` 주소가 보입니다. 그 주소를 쓰세요.

입력:

```
git config --global user.email "내 이메일"
```

### 4-4 편집기 정하기

커밋 메시지를 길게 쓸 때 git이 편집기를 엽니다. 익숙한 편집기가 없다면 nano가 가장 쉽습니다.

입력:

```
git config --global core.editor nano
```

VS Code를 쓴다면 대신 이렇게 입력합니다.

```
git config --global core.editor "code --wait"
```

### 4-5 등록한 값 확인

입력:

```
git config --global --list
```

결과 (예시):

```
credential.https://github.com.helper=
credential.https://github.com.helper=!/usr/bin/gh auth git-credential
user.name=내 이름
user.email=12345678+myid@users.noreply.github.com
core.editor=nano
```

`user.name`, `user.email`, `core.editor` 세 줄이 보이면 됩니다. 순서는 달라도 괜찮습니다.

## 5. fork와 clone

- **fork**: GitHub 안에서 남의 저장소를 내 계정으로 복사합니다. 내 것이므로 마음대로 push할 수 있습니다.
- **clone**: GitHub에 있는 저장소를 내 컴퓨터로 받습니다.

### 5-1 fork

1. 이 저장소의 GitHub 페이지에서 오른쪽 위 **Fork**를 누릅니다.
2. 다음 화면의 **Copy the `main` branch only** 체크를 **해제**합니다. 체크된 채로 만들면 미션 브랜치가 복사되지 않습니다.
3. **Create fork**를 누릅니다.

주소창이 `github.com/<내 아이디>/git-practice`로 바뀌면 성공입니다.

### 5-2 clone

입력:

```
git clone https://github.com/<내 아이디>/git-practice.git
```

결과 (예시, 숫자는 다를 수 있습니다):

```
Cloning into 'git-practice'...
remote: Enumerating objects: 160, done.
Receiving objects: 100% (160/160), done.
Resolving deltas: 100% (52/52), done.
```

### 5-3 받은 폴더로 이동

입력:

```
cd git-practice
```

### 5-4 미션 브랜치가 복사됐는지 확인

입력:

```
git branch -r
```

결과:

```
  origin/HEAD -> origin/main
  origin/main
  origin/mission/m01
  origin/mission/m02
  origin/mission/m03
  origin/mission/m04
  origin/teammate/m04
```

`origin/mission/m01`부터 `origin/mission/m04`까지 보이면 fork가 제대로 된 것입니다. `origin/main`만 보이면 5-1에서 체크 해제를 하지 않은 것입니다. fork를 지우고 다시 만드세요.

### 5-5 수업 자료를 따로 복사해 두기

수업 자료(`lessons` 폴더)는 `main` 브랜치에만 있습니다. 02장부터 미션 브랜치로 옮겨 가면 이 폴더가 사라집니다. 지금 저장소 밖으로 복사해 둡니다.

입력:

```
cp -r lessons ../git-practice-lessons
```

결과는 아무것도 나오지 않습니다. 앞으로 노트북은 저장소 옆에 생긴 `git-practice-lessons` 폴더에서 엽니다. 저장소 안의 `lessons`에서 열면 브랜치를 바꿀 때 노트북이 닫히거나 사라집니다.

## 6. push 권한 확인

### 6-1 받아 온 주소 확인

입력:

```
git remote -v
```

결과 (예시):

```
origin	https://github.com/<내 아이디>/git-practice.git (fetch)
origin	https://github.com/<내 아이디>/git-practice.git (push)
```

주소에 **내 아이디**가 있어야 합니다. 다른 아이디가 있으면 내 fork가 아닌 곳을 clone한 것입니다. 폴더를 지우고(`cd ..` 다음 `rm -rf git-practice`) 5-2를 다시 합니다.

### 6-2 실제로 올리지 않고 push 연습

`--dry-run`은 GitHub에 연결해서 로그인과 권한만 확인하고, 실제로는 아무것도 올리지 않습니다.

입력:

```
git push --dry-run
```

결과:

```
Everything up-to-date
```

이 줄이 나오면 내 fork에 push할 수 있습니다.

- `Username for 'https://github.com':`가 나오면 3-4의 로그인이 git에 연결되지 않은 것입니다. `Ctrl + C`로 멈추고 `gh auth setup-git`을 입력한 뒤 6-2를 다시 합니다.
- `Permission denied` 또는 `403`이 나오면 6-1의 주소가 내 fork가 아닙니다.

## 7. 테스트 실행

### 7-1 kitchen 프로젝트의 테스트 돌리기

입력:

```
python3 -m unittest
```

결과:

```
.........
----------------------------------------------------------------------
Ran 9 tests in 0.001s

OK
```

마지막 줄에 `OK`가 나오면 준비가 끝났습니다. 점(`.`) 하나가 통과한 테스트 하나입니다.

테스트를 돌리면 Python이 `kitchen/__pycache__` 폴더를 만듭니다. 실행 속도를 높이려고 만드는 임시 파일입니다. `.gitignore`에 적혀 있어서 `git status`에는 나오지 않습니다.
