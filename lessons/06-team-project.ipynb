# 06 팀 프로젝트: 여러 사람의 PR을 한 저장소에 모으기

지금까지는 혼자 미션 브랜치를 풀었습니다. 실제 팀에서는 서너 명이 **한 저장소**에 동시에 변경을 올립니다.
같은 파일을 여러 사람이 고치고, 내 PR이 리뷰를 기다리는 사이 다른 사람의 PR이 먼저 합쳐집니다.

이 장에서는 3-4명이 한 팀이 되어 실제로 그 상황을 만듭니다.
팀원 모두가 `kitchen/menu.py`의 **같은 자리**에 메뉴를 하나씩 추가합니다. 그래서 충돌이 반드시 납니다.
02장부터 05장까지 배운 명령을 이 한 번의 작업 안에서 모두 다시 씁니다.

## 따라 하는 방법

- 회색 상자 **입력**의 명령을 터미널에 그대로 입력합니다. 한 번에 한 줄씩 입력합니다.
- 상자 **결과**는 예시 팀에서 실제로 나온 출력입니다. 예시 팀의 팀원은 이렇습니다.

  | 역할 | 이름 | 하는 일 |
  |---|---|---|
  | 나 | Kim Minji | 만두(`mandu`)를 추가합니다 |
  | 팀원 | Lee Jun | 잡채(`japchae`)를 추가합니다 |
  | 팀장 | Park Seoyeon | 팀 저장소를 만들고 PR을 merge합니다 |

- 결과의 `<팀장 아이디>`는 우리 팀 팀장의 GitHub 아이디로 바꿔 읽습니다.
- 이번 장에서 새로 만드는 커밋의 ID는 `???????`로 가렸습니다. 사람마다 이름과 시각이 달라서 내 화면에는 다른 ID와 시각이 나옵니다. 앞 장에서 받은 커밋의 ID는 그대로 같습니다.
- 내 메뉴 이름과 가격은 팀원과 겹치지 않게 정합니다. 결과에 나온 `mandu` 대신 내가 정한 메뉴를 씁니다.

## 0. 팀 준비

### 0-1 팀과 팀장 정하기

3-4명이 한 팀입니다. 그중 한 사람이 팀장을 맡습니다.
팀장의 `git-practice` fork가 팀 저장소가 됩니다. 팀원은 각자 fork를 만들지 않고, 팀장의 저장소 하나에 함께 push합니다.

### 0-2 팀장: 팀원을 Collaborator로 초대

팀장만 합니다.

1. GitHub에서 내 `git-practice` 저장소를 엽니다.
2. **Settings** → 왼쪽 **Collaborators** → **Add people**을 누릅니다.
3. 팀원의 GitHub 아이디를 하나씩 넣고 초대합니다.

초대받은 사람만 이 저장소에 push할 수 있습니다. 초대하지 않은 사람이 push하면 거절됩니다.

### 0-3 팀원: 초대 수락

팀원만 합니다.

GitHub 가입 이메일로 초대 메일이 옵니다. 메일의 **View invitation**을 누르고 **Accept invitation**을 누릅니다.
메일이 보이지 않으면 `https://github.com/<팀장 아이디>/git-practice/invitations` 주소를 직접 엽니다.

> 수락하기 전에는 push가 거절됩니다. 2-6에서 `Permission denied`나 `403`이 나오면 이 단계부터 확인합니다.

### 0-4 팀장: main 보호하기 (선택)

팀 규칙 "main에는 직접 push하지 않는다"를 GitHub가 지키게 할 수 있습니다.

1. **Settings** → **Branches** → **Add branch ruleset** (화면에 따라 **Rules** → **Rulesets**)을 누릅니다.
2. 이름을 `main`으로 짓고, **Enforcement status**를 **Active**로 바꿉니다.
3. **Target branches**에서 **Include default branch**를 고릅니다.
4. **Require a pull request before merging**에 체크하고 저장합니다.

이제 누군가 실수로 `git push origin main`을 해도 GitHub가 거절합니다.

## 2. 내 메뉴를 브랜치에 올리기

팀 규칙은 세 가지입니다.

| 규칙 | 이유 |
|---|---|
| main에는 직접 push하지 않는다 | main은 모두가 받아 가는 브랜치입니다. 검토하지 않은 변경이 바로 모두에게 갑니다 |
| 일 하나에 브랜치 하나, 이름은 `menu-<내 이름>` | 누가 무슨 일을 하는 브랜치인지 이름만 보고 압니다 |
| PR은 팀원 한 명이 승인한 뒤 merge한다 | 내가 놓친 것을 다른 사람이 봅니다 |

### 2-1 브랜치 만들기

입력:

```
git switch -c menu-minji
```

결과:

```
Switched to a new branch 'menu-minji'
```

`minji` 자리에는 내 이름을 씁니다.

### 2-2 메뉴 추가

편집기로 `kitchen/menu.py`를 열고, `"dinner set"` 줄 바로 아래에 내 메뉴를 한 줄 추가합니다. 예시 팀의 Minji는 이렇게 썼습니다.

```python
    "dinner set": 25.0,
    "mandu": 6,
}
```

### 2-3 바뀐 내용 확인

```
git diff
```

## 3. 팀 저장소에 PR 열기

### 3-1 PR 만들기

1. `https://github.com/<팀장 아이디>/git-practice`를 엽니다.
2. 노란 띠의 **Compare & pull request**를 누릅니다. 띠가 없으면 **Pull requests** → **New pull request**를 누릅니다.
3. 화면 위쪽의 네 칸을 확인합니다.

| 칸 | 값 |
|---|---|
| base repository | `<팀장 아이디>/git-practice` |
| base | `main` |
| head repository | `<팀장 아이디>/git-practice` |
| compare | `menu-minji` (내 브랜치) |

4. 제목과 설명을 쓰고 오른쪽 **Reviewers**에서 팀원 한 명을 고른 뒤 **Create pull request**를 누릅니다.

> **base repository를 꼭 확인하세요.** 팀장의 저장소는 `Drillcheck/git-practice`의 fork라서, GitHub가 처음에 base를 `Drillcheck/git-practice`로 골라 두는 경우가 많습니다. 그대로 만들면 PR이 팀 저장소가 아니라 원본 저장소로 갑니다.

## 4. 리뷰 주고받기

### 4-1 팀원의 브랜치 받기

내 PR을 기다리는 동안 나도 팀원의 PR을 리뷰합니다. 예시 팀에서는 Jun이 잡채를 추가한 `menu-jun`을 push했습니다.
다른 사람이 올린 브랜치는 `git fetch`로 받습니다.

입력:

```
git fetch origin
```

결과:

```
From https://github.com/<팀장 아이디>/git-practice
 * [new branch]      menu-jun   -> origin/menu-jun
```

> 00장에서 설치한 GitHub CLI로는 `gh pr checkout 2`처럼 PR 번호만으로 그 브랜치를 받아서 이동할 수도 있습니다. 이 장에서는 git 명령으로 합니다.

### 4-2 팀원의 브랜치에서 테스트

PR 화면의 코드만 읽어서는 테스트가 통과하는지 모릅니다. 내 컴퓨터에서 직접 돌려 봅니다.

입력:

```
git switch menu-jun
python3 -m unittest
```

결과:

```
Switched to a new branch 'menu-jun'
branch 'menu-jun' set up to track 'origin/menu-jun'.
.........
----------------------------------------------------------------------
Ran 9 tests in 0.001s

OK
```

### 4-3 리뷰 남기기

1. 팀원의 PR에서 **Files changed** 탭을 엽니다.
2. 고칠 점이 있으면 그 줄 왼쪽의 **+**를 눌러 의견을 씁니다.
3. 오른쪽 위 **Review changes**에서 하나를 고르고 **Submit review**를 누릅니다.

| 선택 | 뜻 |
|---|---|
| Comment | 의견만 남깁니다 |
| Approve | 이대로 merge해도 된다는 승인입니다 |
| Request changes | 고친 뒤에 다시 보겠다는 뜻입니다 |

다 봤으면 내 브랜치로 돌아옵니다.

입력:

```
git switch menu-minji
```

결과:

```
Switched to branch 'menu-minji'
Your branch is up to date with 'origin/menu-minji'.
```

### 4-4 리뷰 반영: 마지막 커밋 고치기

예시 팀에서는 Jun이 Minji의 PR에 이런 의견을 남겼습니다.

> 다른 메뉴는 가격을 `9.5`, `25.0`처럼 소수점까지 씁니다. `6`도 `6.0`으로 맞춰 주세요.

아직 merge되지 않은 내 PR이고, 고칠 것은 방금 커밋한 그 한 줄입니다. 03장에서 배운 `--amend`로 마지막 커밋을 고칩니다.
편집기에서 `"mandu": 6,`을 `"mandu": 6.0,`으로 고칩니다.

입력:

```
git add kitchen/menu.py
git commit --amend --no-edit
```

결과:

```
[menu-minji ???????] Add mandu to the menu
 Date: Fri Nov 20 10:00:00 2026 +0900
 1 file changed, 1 insertion(+)
```

`--no-edit`는 커밋 메시지를 그대로 둔다는 뜻입니다.

### 4-5 그냥 push하면

입력:

```
git push
```

결과:

```
To https://github.com/<팀장 아이디>/git-practice.git
 ! [rejected]        menu-minji -> menu-minji (non-fast-forward)
error: failed to push some refs to 'https://github.com/<팀장 아이디>/git-practice.git'
hint: Updates were rejected because the tip of your current branch is behind
hint: its remote counterpart. If you want to integrate the remote changes,
hint: use 'git pull' before pushing again.
hint: See the 'Note about fast-forwards' in 'git push --help' for details.
```

amend한 커밋은 새 커밋입니다. GitHub의 `menu-minji`에는 고치기 전 커밋이 있으니, 그대로 덮어쓰면 그 커밋이 사라집니다. 그래서 git이 거절합니다.

### 4-6 --force-with-lease로 push

내 PR 브랜치이고, 고치기 전 커밋이 사라져도 된다는 것을 압니다. 이럴 때 덮어쓰기를 허락하는 옵션이 `--force-with-lease`입니다.

입력:

```
git push --force-with-lease
```

결과:

```
To https://github.com/<팀장 아이디>/git-practice.git
 + ???????...??????? menu-minji -> menu-minji (forced update)
```

`(forced update)`는 덮어썼다는 표시입니다. PR 화면에는 고친 커밋이 바로 반영됩니다.

### 퀴즈 06-3

`--force`도 덮어쓰기를 허락합니다. 리뷰어가 내 PR 브랜치에 오타 수정 커밋을 하나 직접 push해 둔 상태라면, `--force`와 `--force-with-lease`는 각각 어떻게 됩니까?

<details><summary>힌트</summary>

`--force-with-lease`는 "내가 마지막으로 본 GitHub의 상태가 지금도 그대로일 때만" 덮어씁니다. 내가 마지막으로 본 상태는 `git fetch`나 `git push`를 한 시점의 `origin/menu-minji`입니다.

</details>

## 5. 다른 사람의 PR이 먼저 merge됐을 때

### 5-1 GitHub에서 보이는 것

예시 팀에서는 Jun의 PR이 먼저 승인을 받아 merge되었습니다.

그러자 Minji의 PR 아래쪽에 **This branch has conflicts that must be resolved**가 나오고 merge 버튼이 회색이 되었습니다.
Jun과 Minji가 `"dinner set"` 바로 아래, 같은 자리에 줄을 추가했기 때문입니다.
충돌은 GitHub 화면에서 고칠 수도 있지만, 여기서는 내 컴퓨터에서 고칩니다. 고친 다음 테스트를 돌려 볼 수 있기 때문입니다.

### 5-2 팀 저장소의 최신 상태 받기

입력:

```
git fetch origin
```

결과:

```
From https://github.com/<팀장 아이디>/git-practice
   ???????..???????  main       -> origin/main
```

`main`이 앞으로 나갔습니다. 무엇이 새로 들어왔는지 봅니다.

입력:

```
git log --oneline menu-minji..origin/main
```

결과:

```
??????? Merge pull request #2 from <팀장 아이디>/menu-jun
??????? Add japchae to the menu
```

`A..B`는 "B에는 있고 A에는 없는 커밋"입니다. Jun의 커밋과, 그것을 합친 merge 커밋이 보입니다.

### 5-3 최신 main 위로 내 커밋 옮기기

05장에서 배운 `rebase`로 내 커밋을 최신 `origin/main` 위에 다시 쌓습니다.

입력:

```
git rebase origin/main
```

결과:

```
Auto-merging kitchen/menu.py
CONFLICT (content): Merge conflict in kitchen/menu.py
error: could not apply ???????... Add mandu to the menu
hint: Resolve all conflicts manually, mark them as resolved with
hint: "git add/rm <conflicted_files>", then run "git rebase --continue".
hint: You can instead skip this commit: run "git rebase --skip".
hint: To abort and get back to the state before "git rebase", run "git rebase --abort".
hint: Disable this message with "git config set advice.mergeConflict false"
Could not apply ???????... # Add mandu to the menu
```

### 5-4 충돌 난 파일 확인

입력:

```
git status
```

결과:

```
interactive rebase in progress; onto ???????
Last command done (1 command done):
   pick ??????? # Add mandu to the menu
No commands remaining.
You are currently rebasing branch 'menu-minji' on '???????'.
  (fix conflicts and then run "git rebase --continue")
  (use "git rebase --skip" to skip this patch)
  (use "git rebase --abort" to check out the original branch)

Unmerged paths:
  (use "git restore --staged <file>..." to unstage)
  (use "git add <file>..." to mark resolution)
	both modified:   kitchen/menu.py

no changes added to commit (use "git add" and/or "git commit -a")
```

입력:

```
cat kitchen/menu.py
```

결과:

```
MENU = {
    "kimchi stew": 9.5,
    "bibimbap": 10.5,
    "tteokbokki": 7.0,
    "barley tea": 1.5,
    "kimbap": 4.5,
    "dinner set": 25.0,
<<<<<<< HEAD
    "japchae": 8.5,
||||||| parent of ??????? (Add mandu to the menu)
=======
    "mandu": 6.0,
>>>>>>> ??????? (Add mandu to the menu)
}
```

| 부분 | 뜻 |
|---|---|
| `<<<<<<<`부터 `\|\|\|\|\|\|\|`까지 | 옮겨 갈 자리, 즉 `origin/main`에 이미 있는 내용 (Jun의 잡채) |
| `\|\|\|\|\|\|\|`부터 `=======`까지 | 두 사람이 고치기 전의 원래 줄 (05장 8절에서 켠 zdiff3 표시) |
| `=======`부터 `>>>>>>>`까지 | 옮기는 중인 내 커밋의 내용 (Minji의 만두) |

> rebase에서는 `HEAD`가 "옮겨 갈 자리"입니다. merge 때와 위아래의 주인이 반대로 보이니 표시 뒤의 이름을 읽고 판단합니다.

### 5-5 두 메뉴를 모두 살리기

메뉴 둘 다 남아야 합니다. 편집기에서 표시 네 줄을 지우고 두 메뉴 줄을 모두 남깁니다. 원래 줄 부분이 비어 있다면 표시 줄만 지우면 됩니다.

```python
    "dinner set": 25.0,
    "japchae": 8.5,
    "mandu": 6.0,
}
```

### 5-6 고친 파일 표시하고 rebase 계속하기

입력:

```
git add kitchen/menu.py
git rebase --continue
```

결과:

```
[detached HEAD ???????] Add mandu to the menu
 1 file changed, 1 insertion(+)
Successfully rebased and updated refs/heads/menu-minji.
```

커밋 메시지 편집기가 열리면 메시지를 그대로 두고 저장한 뒤 닫습니다. nano라면 `Ctrl+O`, Enter, `Ctrl+X`입니다.

결과:

```
.........
----------------------------------------------------------------------
Ran 9 tests in 0.001s

OK
```

### 5-8 다시 push

rebase도 커밋을 새로 만듭니다. 그래서 4-6처럼 `--force-with-lease`로 push합니다.

입력:

```
git push --force-with-lease
```

결과:

```
To https://github.com/<팀장 아이디>/git-practice.git
 + ???????...??????? menu-minji -> menu-minji (forced update)
```

PR 화면의 충돌 표시가 사라지고 merge 버튼이 다시 초록색이 됩니다. 리뷰어의 승인을 받으면 팀장이 merge합니다.

## 6. merge된 뒤 정리하기

### 6-1 main으로 돌아와 최신으로 맞추기

예시 팀의 팀장이 Minji의 PR도 merge했습니다. 이제 모두가 자기 컴퓨터의 `main`을 최신으로 맞춥니다.

입력:

```
git switch main
```

결과:

```
Switched to branch 'main'
Your branch is behind 'origin/main' by 2 commits, and can be fast-forwarded.
  (use "git pull" to update your local branch)
```

`behind ... by 2 commits`는 5-2에서 받아 둔 Jun의 커밋 두 개입니다. 그 뒤에 merge된 내 PR은 아직 받지 않아서 git도 모릅니다.

입력:

```
git pull
```

결과:

```
From https://github.com/<팀장 아이디>/git-practice
   ???????..???????  main       -> origin/main
Updating ???????..???????
Fast-forward
 kitchen/menu.py | 2 ++
 1 file changed, 2 insertions(+)
```

`git pull`은 `git fetch`와 merge를 한 번에 합니다. 내 `main`에는 따로 커밋한 것이 없으니 그냥 앞으로 당겨집니다(`Fast-forward`).

### 6-4 다 쓴 브랜치 지우기

merge된 브랜치는 지웁니다. 남겨 두면 어느 브랜치가 아직 진행 중인지 알기 어렵습니다.

입력:

```
git branch -d menu-minji
```

결과:

```
Deleted branch menu-minji (was ???????).
```

`-d`는 merge된 브랜치만 지웁니다. 아직 합쳐지지 않은 커밋이 있으면 거절해서 실수로 작업을 잃지 않게 합니다.

GitHub에 있는 브랜치도 지웁니다. PR 화면의 **Delete branch** 버튼을 눌러도 같습니다.

입력:

```
git push origin --delete menu-minji
```

결과:

```
To https://github.com/<팀장 아이디>/git-practice.git
 - [deleted]         menu-minji
```

### 퀴즈 06-6

`git branch -d menu-jun`을 입력하면 어떻게 됩니까? 지워진다면 왜 지워도 되고, 거절된다면 왜 거절됩니까?

<details><summary>힌트</summary>

`menu-jun`은 4-2에서 받은 Jun의 브랜치입니다. Jun의 커밋은 이미 `main`에 들어와 있습니다.

</details>

### 퀴즈 06-7

팀원 한 명이 브랜치를 만들지 않고 `main`에서 커밋해서 `git push origin main`을 했습니다. 0-4의 main 보호를 켠 팀과 켜지 않은 팀에서 각각 어떻게 됩니까? 켜지 않은 팀에서 이미 push가 되었다면, 그 커밋을 없애려고 `main`에 `--force-with-lease`를 써도 됩니까?

### 이 과정의 명령, 상황별로

| 상황 | 명령 | 처음 나온 장 |
|---|---|---|
| 이력 읽기 | `git log --oneline --graph`, `git show`, `git blame` | 01 |
| 작업 시작 | `git switch -c 브랜치` | 02 |
| 무엇이 바뀌었나 | `git status`, `git diff` | 02 |
| 커밋 | `git add 파일`, `git commit -m "메시지"` | 02 |
| 처음 push | `git push -u origin 브랜치` | 02 |
| 커밋 전 변경 되돌리기 | `git restore 파일`, `git restore --staged 파일` | 03 |
| 하던 일 잠시 치우기 | `git stash`, `git stash pop` | 03 |
| 마지막 커밋 고치기 | `git commit --amend` | 03 |
| 커밋 여러 개 다듬기 | `git rebase -i` | 04 |
| 최신 변경 받기 | `git fetch origin` | 05 |
| 최신 main 위로 옮기기 | `git rebase origin/main` | 05 |
| 충돌 풀기 | 파일 고치기, `git add`, `git rebase --continue` 또는 `git commit` | 05 |
| 고친 커밋 다시 push | `git push --force-with-lease` | 05 |
| merge 뒤 정리 | `git switch main`, `git pull`, `git branch -d`, `git push origin --delete` | 06 |

### 실제 팀 프로젝트를 시작할 때

- [ ] 팀 저장소 하나를 정하고, 팀원을 Collaborator로 초대했다
- [ ] main 보호를 켜서 PR 없이는 main에 들어가지 않게 했다
- [ ] 브랜치 이름 규칙을 정했다 (예: `기능-이름`)
- [ ] PR마다 리뷰어 한 명 이상이 승인한 뒤 merge한다
- [ ] PR을 열 때 base repository와 base 브랜치를 확인한다
- [ ] 작업을 시작하기 전, PR을 올리기 전에 `git fetch`로 최신 main을 받는다
- [ ] `--force-with-lease`는 내 PR 브랜치에만 쓰고, main에는 쓰지 않는다
- [ ] merge된 브랜치는 지운다
