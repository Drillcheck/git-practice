# 03 실수 되돌리기

git을 쓰다 보면 이런 순간이 옵니다.

- 파일을 고치다 망쳤는데, 아직 커밋하지 않았다.
- `git add`로 올리지 말아야 할 파일까지 올렸다.
- 하던 일을 잠깐 치워 두고 다른 브랜치를 봐야 한다.
- 방금 만든 커밋을 없던 일로 하고 싶다.
- 마지막 커밋에 파일 하나를 빠뜨렸다.
- 커밋에 비밀 키가 들어갔다.
- 이미 push한 커밋을 되돌려야 한다.

상황마다 쓰는 명령이 다릅니다. 이 장에서는 연습용 브랜치에서 하나씩 실수를 만들고 되돌려 본 다음, 미션 m02와 m03을 풉니다.

## 따라 하는 방법

- 회색 상자 **입력**의 명령을 터미널에 그대로 입력하고 Enter를 누릅니다. 한 번에 한 줄씩 입력합니다.
- 상자 **결과**는 이 저장소에서 실제로 나오는 출력입니다. 내 화면과 비교해 보세요.
- **내가 만든 커밋의 ID는 `???????`로 가렸습니다.** 커밋 ID는 작성자와 시각까지 넣어 계산하기 때문에, 내 화면에는 다른 ID와 시각이 나옵니다. 처음부터 저장소에 있던 커밋은 ID까지 똑같이 나옵니다.
- "편집기로 엽니다"라고 적힌 단계는 VS Code 같은 편집기에서 파일을 고치고 저장합니다.
- Windows에서 `python3`이 없다고 나오면 `python`으로 입력합니다.

## 1. 저장하기 전 파일을 망쳤다

### 1-1 파일 망가뜨리기

`kitchen/menu.py`를 편집기로 엽니다. `"kimchi stew": 9.0,` 줄을 지우고 저장합니다.

입력:

```
git status
```

결과:

```
On branch practice
Your branch is up to date with 'origin/mission/m01'.

Changes not staged for commit:
  (use "git add <file>..." to update what will be committed)
  (use "git restore <file>..." to discard changes in working directory)
	modified:   kitchen/menu.py

no changes added to commit (use "git add" and/or "git commit -a")
```

`modified:` 아래에 고친 파일이 보입니다. 아직 `git add`도 커밋도 하지 않은 상태입니다.

### 1-2 무엇이 바뀌었나 보기

입력:

```
git diff
```

결과:

```
diff --git a/kitchen/menu.py b/kitchen/menu.py
index 4e31d33..ea88d22 100644
--- a/kitchen/menu.py
+++ b/kitchen/menu.py
@@ -1,5 +1,4 @@
 MENU = {
-    "kimchi stew": 9.0,
     "bibimbap": 10.5,
     "tteokbokki": 7.0,
     "barley tea": 1.5,
```

`-`로 시작하는 줄이 지워진 줄입니다. 실수로 지운 줄을 하나씩 다시 써 넣을 수도 있지만, 바뀐 곳이 많으면 어디를 고쳤는지 기억나지 않습니다.

### 1-3 마지막 커밋 상태로 되돌리기

입력:

```
git restore kitchen/menu.py
git status
```

결과:

```
On branch practice
Your branch is up to date with 'origin/mission/m01'.

nothing to commit, working tree clean
```

`git restore 파일`은 그 파일을 마지막 커밋 때 내용으로 덮어씁니다. 편집기에서 다시 열어 보면 지운 줄이 돌아와 있습니다.

> `git restore`로 덮어쓴 변경은 git 어디에도 남지 않습니다. 한 번도 `git add`나 커밋을 하지 않은 내용이기 때문입니다. 되돌리기 전에 `git diff`로 버려도 되는 변경인지 꼭 확인합니다.

## 2. add하지 말아야 할 파일까지 add했다

### 2-1 두 파일을 한꺼번에 올리기

메뉴에 김밥을 추가합니다. `kitchen/menu.py`를 편집기로 열어 `"barley tea": 1.5,` 아래에 다음 줄을 넣고 저장합니다.

```python
    "kimbap": 4.5,
```

그리고 작업하면서 적어 둔 메모 파일 `notes.txt`를 저장소 폴더에 만듭니다. 내용은 아무거나 괜찮습니다.

입력:

```
git add .
git status
```

결과:

```
On branch practice
Your branch is up to date with 'origin/mission/m01'.

Changes to be committed:
  (use "git restore --staged <file>..." to unstage)
	modified:   kitchen/menu.py
	new file:   notes.txt
```

`git add .`은 "이 폴더의 바뀐 파일 전부"라는 뜻입니다. 그래서 커밋에 넣을 생각이 없던 `notes.txt`까지 `Changes to be committed:`에 올라갔습니다.
이 상태에서 커밋하면 메모가 팀 저장소에 들어갑니다.

### 2-2 한 파일만 내리기

입력:

```
git restore --staged notes.txt
git status
```

결과:

```
On branch practice
Your branch is up to date with 'origin/mission/m01'.

Changes to be committed:
  (use "git restore --staged <file>..." to unstage)
	modified:   kitchen/menu.py

Untracked files:
  (use "git add <file>..." to include in what will be committed)
	notes.txt
```

`--staged`를 붙이면 파일 내용은 그대로 두고 **커밋에 넣을 목록에서만** 뺍니다. `notes.txt`는 지워지지 않았고, `Untracked files:`로 돌아갔습니다.
`git status`의 안내 문구에도 이 명령이 적혀 있습니다. 막히면 `git status`부터 읽어 보세요.

### 퀴즈 03-2

## 3. 하던 일을 잠깐 치워 두기

### 3-1 치워 두기

김밥을 넣는 중에 팀원이 다른 브랜치의 코드를 같이 봐 달라고 합니다. 지금 변경은 아직 커밋할 만큼 끝나지 않았습니다.
이럴 때 변경을 잠깐 다른 곳에 넣어 두는 명령이 `git stash`입니다.

입력:

```
git stash
git status
```

결과:

```
Saved working directory and index state WIP on practice: 5c49179 Add the order total
On branch practice
Your branch is up to date with 'origin/mission/m01'.

Untracked files:
  (use "git add <file>..." to include in what will be committed)
	notes.txt

nothing added to commit but untracked files present (use "git add" to track)
```

`kitchen/menu.py`의 변경이 사라지고 폴더가 마지막 커밋 상태로 돌아갔습니다. 이제 다른 브랜치로 옮겨 가도 내 변경이 따라오지 않습니다.
그런데 `notes.txt`는 그대로 남아 있습니다. stash는 git이 이미 알고 있는 파일의 변경만 넣어 둡니다. 한 번도 add하지 않은 새 파일(`Untracked`)은 넣지 않습니다.

### 3-2 넣어 둔 목록 보기

입력:

```
git stash list
```

결과:

```
stash@{0}: WIP on practice: 5c49179 Add the order total
```

`stash@{0}`이 방금 넣어 둔 것입니다. 여러 번 stash하면 `stash@{1}`, `stash@{2}`로 쌓입니다. 끝의 메시지는 넣어 둘 때의 브랜치와 마지막 커밋입니다.

### 3-3 다시 꺼내기

다른 브랜치 일을 마치고 돌아왔다고 하겠습니다.

입력:

```
git stash pop
```

결과:

```
On branch practice
Your branch is up to date with 'origin/mission/m01'.

Changes not staged for commit:
  (use "git add <file>..." to update what will be committed)
  (use "git restore <file>..." to discard changes in working directory)
	modified:   kitchen/menu.py

Untracked files:
  (use "git add <file>..." to include in what will be committed)
	notes.txt

no changes added to commit (use "git add" and/or "git commit -a")
Dropped refs/stash@{0} (????????????????????????????????????????)
```

`pop`은 넣어 둔 변경을 다시 꺼내고 목록에서 지웁니다. 김밥 줄이 돌아왔습니다.

### 3-4 김밥 커밋하기

메모 파일은 지우고, 김밥만 커밋합니다.

입력:

```
rm notes.txt
git add kitchen/menu.py
git commit -m "Add kimbap to the menu"
```

결과:

```
[practice ???????] Add kimbap to the menu
 1 file changed, 1 insertion(+)
```

## 4. 방금 만든 커밋을 없던 일로 하고 싶다

### 4-1 커밋만 취소하기

김밥 커밋을 만들고 보니, 가격을 Daniel에게 먼저 물어봐야 했습니다. 커밋은 취소하되, 고친 내용은 남겨 두고 싶습니다.

입력:

```
git reset --soft HEAD~1
git status
```

결과:

```
On branch practice
Your branch is up to date with 'origin/mission/m01'.

Changes to be committed:
  (use "git restore --staged <file>..." to unstage)
	modified:   kitchen/menu.py
```

| 부분 | 뜻 |
|---|---|
| `HEAD` | 지금 내가 있는 커밋. 방금 만든 김밥 커밋입니다 |
| `HEAD~1` | 그 하나 전 커밋 |
| `git reset HEAD~1` | 브랜치를 하나 전 커밋으로 옮깁니다. 김밥 커밋이 브랜치에서 빠집니다 |
| `--soft` | 파일과 add한 목록은 건드리지 않습니다 |

김밥 줄은 `Changes to be committed:`에 그대로 있습니다. 바로 다시 커밋할 수 있는 상태입니다.

입력:

```
git log --oneline -3
```

결과:

```
5c49179 Add the order total
81570c5 Start the kitchen order calculator
e71d2c2 Add mission m01 guide
```

맨 위가 다시 `Add the order total`입니다.

### 4-2 add까지 취소하기

같은 커밋을 다시 만든 다음, 이번에는 `--soft` 없이 취소해 봅니다.

입력:

```
git commit -m "Add kimbap to the menu"
git reset HEAD~1
git status
```

결과:

```
[practice ???????] Add kimbap to the menu
 1 file changed, 1 insertion(+)
Unstaged changes after reset:
M	kitchen/menu.py
On branch practice
Your branch is up to date with 'origin/mission/m01'.

Changes not staged for commit:
  (use "git add <file>..." to update what will be committed)
  (use "git restore <file>..." to discard changes in working directory)
	modified:   kitchen/menu.py

no changes added to commit (use "git add" and/or "git commit -a")
```

옵션 없이 쓰면 커밋을 취소하고 add한 목록도 비웁니다. 파일 내용은 여전히 그대로입니다. 김밥 줄은 이제 `Changes not staged for commit:`에 있습니다.
이 상태에서 다시 커밋하려면 `git add`부터 합니다.

### 4-3 --hard는 파일까지 되돌립니다

인터넷 자료에서는 `git reset --hard`가 자주 나옵니다. 이 옵션은 커밋과 add 목록에 더해 **내 폴더의 파일까지** 그 커밋 때로 덮어씁니다.
직접 겪어 봅니다. 김밥을 다시 커밋하고, 그 위에 커밋하지 않은 변경을 하나 만듭니다. `README.md`를 편집기로 열어 맨 아래에 아무 줄이나 쓰고 저장합니다.

입력:

```
git add kitchen/menu.py
git commit -m "Add kimbap to the menu"
git status --short
```

결과:

```
[practice ???????] Add kimbap to the menu
 1 file changed, 1 insertion(+)
 M README.md
```

`--short`는 한 줄 요약입니다. `M`은 고쳤지만 커밋하지 않은 파일입니다.

입력:

```
git reset --hard HEAD~1
git status
```

결과:

```
HEAD is now at 5c49179 Add the order total
On branch practice
Your branch is up to date with 'origin/mission/m01'.

nothing to commit, working tree clean
```

김밥 커밋도, `README.md`에 쓴 줄도 사라졌습니다.

### 4-4 지운 커밋 되찾기

커밋은 브랜치에서 빠져도 한동안 저장소 안에 남아 있습니다. `git reflog`는 HEAD가 지나온 자리를 모두 보여 줍니다.

입력:

```
git reflog -5
```

결과:

```
5c49179 HEAD@{0}: reset: moving to HEAD~1
??????? HEAD@{1}: commit: Add kimbap to the menu
5c49179 HEAD@{2}: reset: moving to HEAD~1
??????? HEAD@{3}: commit: Add kimbap to the menu
5c49179 HEAD@{4}: reset: moving to HEAD~1
```

`HEAD@{0}`이 지금, `HEAD@{1}`이 바로 전 자리입니다. 바로 전 자리가 김밥 커밋입니다.

입력:

```
git reset --hard HEAD@{1}
git log --oneline -2
```

결과:

```
HEAD is now at ??????? Add kimbap to the menu
??????? Add kimbap to the menu
5c49179 Add the order total
```

김밥 커밋이 돌아왔습니다. 하지만 `README.md`에 쓴 줄은 돌아오지 않습니다. 그 줄은 한 번도 커밋하지 않아서 reflog에도 없습니다.

> `--hard`를 쓰기 전에는 `git status`로 커밋하지 않은 변경이 없는지 먼저 봅니다. 남은 변경이 있으면 커밋하거나 stash한 다음에 씁니다.

### 퀴즈 03-4

|---|---|---|
| `--soft` | | |
| 옵션 없음 | | |
| `--hard` | | |

4-4에서 `git reflog`로 되찾을 수 있는 것과 되찾을 수 없는 것은 각각 무엇이었습니까? 차이는 어디서 생깁니까?

## 5. 마지막 커밋을 고치고 싶다

### 5-1 메시지 고치기

김밥 커밋의 메시지에 가격 정보를 넣고 싶습니다. 커밋을 하나 더 만들지 않고 마지막 커밋 자체를 고칩니다.

입력:

```
git log --oneline -1
git commit --amend -m "Add kimbap to the menu at 4.5"
git log --oneline -2
```

결과:

```
??????? Add kimbap to the menu
[practice ???????] Add kimbap to the menu at 4.5
 Date: Fri Nov 20 10:00:00 2026 +0900
 1 file changed, 1 insertion(+)
??????? Add kimbap to the menu at 4.5
5c49179 Add the order total
```

커밋은 여전히 하나입니다. 메시지만 바뀌었습니다. 내 화면에서 amend 전과 후의 커밋 ID를 비교해 보세요. **ID가 바뀌었습니다.**
amend는 마지막 커밋을 고치는 것이 아니라, 고친 내용으로 새 커밋을 만들어 그 자리에 바꿔 끼웁니다. 원래 커밋은 브랜치에서 빠집니다.

### 5-2 이미 push한 커밋이라면

push한 커밋을 amend하면 내 컴퓨터와 GitHub의 이력이 서로 달라집니다. 그 상태에서 `git push`를 하면 GitHub가 거절합니다.
이때 쓰는 `git push --force-with-lease`는 05장에서 배웁니다. 그 전까지는 **push하기 전의 커밋만 amend한다**고 기억해 두세요.

### 퀴즈 03-6

메시지는 그대로 두고 빠뜨린 파일만 마지막 커밋에 넣고 싶을 때, `-m "..."` 대신 무엇을 붙입니까? `git commit --help`에서 `--amend` 근처를 읽어 보세요.

## 8. 이미 공유한 커밋을 되돌리기

지금까지 쓴 `reset`과 `amend`는 이력을 고칩니다. 아직 나만 가진 커밋이면 괜찮습니다. 그런데 이미 push해서 팀원이 받아 간 커밋을 지우면, 팀원의 이력과 GitHub의 이력이 어긋납니다(05장에서 직접 겪습니다).
이럴 때는 커밋을 지우지 않고, **그 변경을 거꾸로 하는 커밋을 하나 더 쌓습니다.** 이 명령이 `git revert`입니다.

### 8-1 연습용 브랜치로 돌아가기

`practice` 브랜치의 김밥 커밋을 이미 팀과 나눴다고 하겠습니다. 그런데 김밥은 메뉴에서 빼기로 했습니다.

입력:

```
git switch practice
git log --oneline -3
```

결과:

```
Switched to branch 'practice'
Your branch is ahead of 'origin/mission/m01' by 1 commit.
  (use "git push" to publish your local commits)
??????? Add kimbap to the menu at 4.5
5c49179 Add the order total
81570c5 Start the kitchen order calculator
```

### 8-2 김밥 커밋 되돌리기

입력:

```
git revert HEAD
```

편집기가 열리면 `Revert "Add kimbap to the menu at 4.5"`라는 메시지가 이미 적혀 있습니다. 그대로 저장하고 닫습니다.

## 정리

| 상황 | 명령 |
|---|---|
| 커밋 전 파일 변경을 버리기 | `git restore 파일` |
| add한 파일을 목록에서만 내리기 | `git restore --staged 파일` |
| 하던 변경을 잠깐 치워 두기 | `git stash`, 꺼낼 때 `git stash pop` |
| 치워 둔 목록 | `git stash list` |
| 마지막 커밋 취소, 변경은 staged로 | `git reset --soft HEAD~1` |
| 마지막 커밋 취소, 변경은 파일에만 | `git reset HEAD~1` |
| 커밋과 파일 변경까지 모두 버리기 | `git reset --hard HEAD~1` (먼저 `git status`) |
| 지운 커밋 되찾기 | `git reflog`, `git reset --hard HEAD@{번호}` |
| 마지막 커밋 메시지 고치기 | `git commit --amend -m "새 메시지"` |
| 마지막 커밋에 파일 더 넣기 | `git add 파일`, `git commit --amend --no-edit` |
| 파일을 저장소에서만 빼기 | `git rm --cached 파일` |
| 다시 add되지 않게 막기 | `.gitignore`에 이름 적기 |
| 이미 공유한 커밋 되돌리기 | `git revert 커밋` (되돌리는 새 커밋) |
| 파일 하나만 예전 커밋 상태로 | `git restore --source=커밋 파일` |
