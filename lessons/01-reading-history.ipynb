# 01 처음 받은 프로젝트의 이력 읽기

팀에 새로 들어가면 코드를 고치기 전에 이력부터 읽습니다.
이력은 이런 질문에 답해 줍니다.

- 이 프로젝트는 누가 주로 만들었나?
- 최근에 무엇이 바뀌었나?
- 이 줄은 왜 이렇게 생겼고, 누구에게 물어보면 되나?

이 장에서는 코드를 한 줄도 고치지 않습니다. 명령으로 이력을 읽기만 합니다.

## 따라 하는 방법

- 회색 상자 **입력**의 명령을 터미널에 그대로 입력하고 Enter를 누릅니다. 한 번에 한 줄씩 입력합니다.
- 상자 **결과**는 이 저장소에서 실제로 나오는 출력입니다. 내 화면과 비교해 보세요. 커밋 ID까지 똑같이 나와야 정상입니다.
- 결과가 길면 앞부분만 실었습니다.
- **퀴즈**는 답을 싣지 않았습니다. 배운 명령으로 직접 찾아서 적어 두세요.

> 결과에 `???????`로 적힌 커밋 ID가 있습니다. 맨 위 커밋에는 이 수업 자료가 들어 있어서, 자료를 고칠 때마다 그 ID가 바뀝니다. 내 화면에는 실제 ID가 나옵니다.

## 시작하기

### 0-1 저장소 폴더로 이동

00장에서 clone한 폴더로 들어갑니다.

입력:

```
cd git-practice
```

### 0-2 main 브랜치로 이동

이 장의 실습은 모두 `main` 브랜치에서 합니다.

입력:

```
git switch main
```

결과:

```
Already on 'main'
Your branch is up to date with 'origin/main'.
```

이미 `main`에 있으면 위처럼 나옵니다. 다른 브랜치에 있었다면 `Switched to branch 'main'`이 나옵니다.

## 1. 커밋 목록 보기

커밋은 "이 시점의 파일 상태를 저장해 둔 것"입니다. 커밋마다 누가, 언제, 왜 바꿨는지가 함께 남습니다.

### 1-1 git log

입력:

```
git log
```

결과 (앞의 3개):

```
commit ????????????????????????????????????????
Author: DrillCheck <guide@example.invalid>
Date:   Mon Nov 2 09:00:00 2026 +0900

    Add the practice guide and grader

commit 1bddc8dbb8217289b634776e241519052139f4e0
Author: Daniel <daniel@example.invalid>
Date:   Tue Oct 20 10:00:00 2026 +0900

    Add a note about the weekend dinner set

commit 0a4f86aa4f68153b643b62ae03507475fa319e1f
Author: Sophia <sophia@example.invalid>
Date:   Tue Oct 6 12:30:00 2026 +0900

    Extend lunch hours to 15:00
```

커밋 하나가 네 부분으로 되어 있습니다.

| 줄 | 뜻 |
|---|---|
| `commit` 뒤의 40자리 | 커밋 ID. 커밋 내용으로 계산한 값이라 같은 커밋이면 어느 컴퓨터에서나 같습니다 |
| `Author` | 작성자 이름과 이메일 |
| `Date` | 작성 시각. 끝의 `+0900`은 한국 시간이라는 뜻입니다 |
| 한 칸 들여 쓴 줄 | 커밋 메시지. 무엇을 왜 바꿨는지 |

> **화면이 멈춘 것 같나요?** 출력이 화면보다 길면 git이 한 화면씩 보여 줍니다. 화면 맨 아래에 `:`가 보이면 그 상태입니다.
> 스페이스바는 다음 화면, `q`는 빠져나오기입니다. **지금 `q`를 눌러 빠져나오세요.**

### 1-2 한 줄씩 짧게 보기

커밋이 많으면 한 줄에 하나씩 보는 게 편합니다.

입력:

```
git log --oneline
```

결과 (앞의 8개):

```
??????? Add the practice guide and grader
1bddc8d Add a note about the weekend dinner set
0a4f86a Extend lunch hours to 15:00
b4869ae Prepare the 1.0 release
ab20898 Handle an empty order
4c0a478 Sort the receipt lines by dish name
61577f6 Mention the service charge in the README
c1bf54e Print the discount on the receipt
```

앞의 7자리는 커밋 ID의 앞부분입니다. 40자리를 다 쓰지 않아도 대부분의 명령에서 이 7자리로 커밋을 가리킬 수 있습니다.

### 1-3 갈라지고 합쳐진 모양 보기

입력:

```
git log --oneline --graph
```

결과 (앞의 14줄):

```
* ??????? Add the practice guide and grader
* 1bddc8d Add a note about the weekend dinner set
* 0a4f86a Extend lunch hours to 15:00
* b4869ae Prepare the 1.0 release
* ab20898 Handle an empty order
* 4c0a478 Sort the receipt lines by dish name
* 61577f6 Mention the service charge in the README
* c1bf54e Print the discount on the receipt
* aa3926e Change the lunch discount to 15%
*   3ea5c34 Merge branch 'split-bill'
|\  
| * 7c71ff7 Round each share to cents
| * f61e2de Split the bill between guests
|/
```

| 기호 | 뜻 |
|---|---|
| `*` | 커밋 하나 |
| `|\` | 여기서 다른 브랜치가 합쳐짐 (아래에서 위로 읽습니다) |
| `|/` | 여기서 브랜치가 갈라져 나감 |

오른쪽 줄의 커밋들은 다른 브랜치에서 따로 작업한 것입니다. 4절에서 자세히 봅니다.

### 1-4 한 사람의 커밋만 보기

`--author`로 작성자를 지정합니다.

입력:

```
git log --oneline --author=Noah
```

결과:

```
ab20898 Handle an empty order
7c71ff7 Round each share to cents
f61e2de Split the bill between guests
e6c5913 Test when lunch hours end
56204dd Fix the receipt width
```

### 1-5 줄 수 세기

`|`(파이프)는 왼쪽 명령의 출력을 오른쪽 명령에 넘깁니다. `wc -l`은 줄 수를 셉니다.
`git log --oneline`은 커밋 하나가 한 줄이므로, 줄 수가 곧 커밋 수입니다.

입력:

```
git log --oneline --author=Noah | wc -l
```

결과:

```
5
```

macOS에서는 숫자 앞에 빈칸이 붙어 나올 수 있습니다. 같은 뜻입니다.

### 1-6 오래된 것부터 보기

`git log`는 최신 커밋이 맨 위입니다. `--reverse`를 붙이면 순서가 뒤집혀 가장 오래된 커밋이 맨 위로 옵니다.
`head -1`은 맨 앞 한 줄만 보여 줍니다.

입력:

```
git log --oneline --reverse --author=Noah | head -1
```

결과:

```
56204dd Fix the receipt width
```

Noah의 첫 커밋입니다.

## 3. 커밋 하나 들여다보기

### 3-1 메시지로 커밋 찾기

`--grep`은 커밋 메시지에 그 글자가 들어 있는 커밋만 보여 줍니다.

입력:

```
git log --oneline --grep kimbap
```

결과:

```
153c8bc Add kimbap to the menu
```

### 3-2 그 커밋에서 무엇이 바뀌었나

찾은 커밋 ID를 `git show`에 넣습니다.

입력:

```
git show 153c8bc
```

결과:

```
commit 153c8bc5619d983376b13365ea636241680ae8c7
Author: Daniel <daniel@example.invalid>
Date:   Tue May 19 10:00:00 2026 +0900

    Add kimbap to the menu

diff --git a/kitchen/menu.py b/kitchen/menu.py
index 4e31d33..53d62d3 100644
--- a/kitchen/menu.py
+++ b/kitchen/menu.py
@@ -3,4 +3,5 @@ MENU = {
     "bibimbap": 10.5,
     "tteokbokki": 7.0,
     "barley tea": 1.5,
+    "kimbap": 4.5,
 }
```

`diff --git` 아래가 바뀐 내용입니다.

| 표시 | 뜻 |
|---|---|
| `--- a/...`, `+++ b/...` | 바뀌기 전(a)과 후(b)의 파일 |
| `@@ -3,4 +3,5 @@` | 바뀐 위치. 원래 3번째 줄부터 4줄이, 바뀐 뒤 3번째 줄부터 5줄이 됐다는 뜻 |
| `+`로 시작하는 줄 | 더해진 줄 |
| `-`로 시작하는 줄 | 지워진 줄 |
| 기호 없는 줄 | 위치를 알려 주려고 함께 보여 주는 주변 줄 |

김밥 한 줄이 메뉴에 더해졌습니다.

### 3-3 바뀐 파일 목록만 보기

입력:

```
git show --stat 153c8bc
```

결과:

```
commit 153c8bc5619d983376b13365ea636241680ae8c7
Author: Daniel <daniel@example.invalid>
Date:   Tue May 19 10:00:00 2026 +0900

    Add kimbap to the menu

 kitchen/menu.py | 1 +
 1 file changed, 1 insertion(+)
```

`| 1 +`는 이 파일에서 1줄이 더해졌다는 뜻입니다.

## 7. 이 줄은 누가 썼나

### 7-1 줄마다 마지막으로 바꾼 커밋 보기

입력:

```
git blame kitchen/split.py
```

결과:

```
f61e2de (Noah 2026-07-07 20:10:00 +0900 1) def split(amount, guests):
7c71ff7 (Noah 2026-07-09 21:30:00 +0900 2)     share = round(amount / guests, 2)
7c71ff7 (Noah 2026-07-09 21:30:00 +0900 3)     return [share] * guests
```

한 줄에 하나씩, 왼쪽부터 이렇게 읽습니다.

| 부분 | 뜻 |
|---|---|
| `f61e2de` | 이 줄을 마지막으로 바꾼 커밋 ID |
| `Noah` | 그 커밋의 작성자 |
| `2026-07-07 20:10:00 +0900` | 그 커밋의 시각 |
| `1)` | 줄 번호 |
| 나머지 | 지금 파일의 그 줄 내용 |

1번 줄은 7월 7일, 2-3번 줄은 7월 9일에 마지막으로 바뀌었습니다. 궁금한 줄이 있으면 그 커밋 ID를 `git show`에 넣어 이유를 봅니다.

### 7-2 어떤 코드가 처음 생긴 커밋 찾기

blame은 **마지막으로** 바꾼 커밋만 보여 줍니다. 그 줄을 처음 쓴 커밋은 다를 수 있습니다.
어떤 글자가 처음 생긴(또는 사라진) 커밋을 찾을 때는 `-S`를 씁니다.

입력:

```
git log -S "def split" --oneline -- kitchen/
```

결과:

```
f61e2de Split the bill between guests
```

`def split`이라는 글자는 이 커밋에서 처음 생겼습니다.

그 커밋이 정말 주문 합계 계산 코드를 처음 쓴 커밋입니까?

<details><summary>힌트</summary>

찾은 커밋을 `git show --stat`으로 열어 보세요. `=>`가 보이면 무슨 뜻일까요?
그다음 `--follow`를 붙여 다시 찾아보세요.

```
git log --oneline --follow -- kitchen/order.py
```

</details>

## 정리

| 알고 싶은 것 | 명령 |
|---|---|
| 커밋 목록 | `git log --oneline` |
| 갈라지고 합쳐진 모양 | `git log --oneline --graph` |
| 오래된 것부터 | `git log --oneline --reverse` |
| 한 사람의 커밋 | `git log --oneline --author=이름` |
| 개수 세기 | 명령 끝에 `wc -l`을 파이프로 연결 |
| 누가 많이 했나 | `git shortlog -sn` |
| 메시지로 찾기 | `git log --oneline --grep 글자` |
| 이 커밋에서 무엇이 바뀌었나 | `git show <ID>`, 파일 목록만은 `--stat` |
| merge 커밋만 | `git log --merges --oneline` |
| 파일·폴더로 좁히기 | `git log --oneline -- 경로` |
| 기간으로 좁히기 | `git log --oneline --since=날짜 --until=날짜` |
| 옛 버전 구경 | `git switch --detach 태그`, 돌아올 때 `git switch main` |
| 이 줄을 마지막으로 바꾼 커밋 | `git blame 파일` |
| 이 코드가 처음 생긴 커밋 | `git log -S "글자" --oneline -- 폴더/` |
| 옮겨진 파일의 진짜 시작 | `git log --oneline --follow -- 파일` |
