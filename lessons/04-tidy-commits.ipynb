# 04 리뷰 요청대로 커밋 다듬기

README 오타를 고쳐 Pull Request를 올렸더니 리뷰가 달렸습니다.

> 커밋이 너무 지저분해요. `fix typo`와 `oops`는 하나로 합치고, 메시지는 무엇을 고쳤는지 알 수 있게 바꿔 주세요. `wip` 커밋의 `notes.txt`는 저장소에 들어가면 안 됩니다.

코드는 이미 맞습니다. 고칠 것은 **커밋의 모양**입니다.
03장의 `git commit --amend`는 마지막 커밋 하나만 고칩니다. 이번에는 커밋 여러 개를 합치고, 메시지를 바꾸고, 하나는 버려야 합니다.
이럴 때 쓰는 명령이 `git rebase -i`입니다.

## 따라 하는 방법

- 회색 상자 **입력**의 명령을 터미널에 그대로 입력하고 Enter를 누릅니다. 한 번에 한 줄씩 입력합니다.
- 상자 **결과**는 실제로 나오는 출력입니다. 내 화면과 비교해 보세요.
- 이 장에서는 내가 커밋을 만듭니다. 내가 만든 커밋의 ID는 `???????`로 가렸습니다. 내 화면에는 다른 ID와 시각이 나옵니다. 미션 브랜치에 원래 있던 커밋의 ID는 똑같이 나와야 정상입니다.
- "편집기로 고칩니다"라고 적힌 단계는 VS Code 같은 편집기로 파일을 열어 고치고 저장합니다.
- **퀴즈**는 답을 싣지 않았습니다. 직접 해 보고 결과를 적어 두세요.

## 1. 지저분한 커밋 만들기

리뷰를 받은 상태를 직접 만듭니다. 실제로 이렇게 커밋하는 일이 흔합니다.

### 1-1 미션 브랜치에서 새 브랜치 만들기

입력:

```
cd git-practice
git switch -c m01-messy origin/mission/m01
```

결과:

```
Switched to a new branch 'm01-messy'
branch 'm01-messy' set up to track 'origin/mission/m01'.
```

### 1-2 첫 커밋: 오타를 고쳤는데 또 틀림

편집기로 `README.md`를 엽니다. 마지막 줄의 `recieve`를 `receve`로 고칩니다. 일부러 또 틀리게 씁니다.

입력:

```
git commit -am "fix typo"
```

결과:

```
[m01-messy ???????] fix typo
 1 file changed, 1 insertion(+), 1 deletion(-)
```

`-a`는 이미 저장소에 있는 파일 중 바뀐 것을 모두 `git add`한 뒤 커밋합니다. 새로 만든 파일은 넣지 않습니다.

### 1-3 둘째 커밋: 상관없는 내용 추가

`README.md`의 셋째 줄 `A tiny order calculator for a small restaurant.` 바로 아래에 다음 줄을 넣고 저장합니다.

```
Open daily from 11:00 to 21:00.
```

입력:

```
git commit -am "Add a note about opening hours"
```

결과:

```
[m01-messy ???????] Add a note about opening hours
 1 file changed, 1 insertion(+)
```

### 1-4 셋째 커밋: 다시 고친 오타

마지막 줄의 `receve`를 `receive`로 고치고 저장합니다.

입력:

```
git commit -am "oops"
```

결과:

```
[m01-messy ???????] oops
 1 file changed, 1 insertion(+), 1 deletion(-)
```

### 1-5 넷째 커밋: 메모 파일

저장소 맨 위 폴더에 `notes.txt`를 만들고 아래 한 줄을 적어 저장합니다.

```
ask Emily about the receipt width
```

입력:

```
git add notes.txt
git commit -m "wip"
```

결과:

```
[m01-messy ???????] wip
 1 file changed, 1 insertion(+)
 create mode 100644 notes.txt
```

`wip`는 work in progress, "작업 중"의 줄임말입니다. 자기만 알아보는 중간 저장에 흔히 붙입니다.

### 1-6 지금 모양 보기

입력:

```
git log --oneline
```

결과:

```
??????? wip
??????? oops
??????? Add a note about opening hours
??????? fix typo
6389ba2 Add the order total
a3ec238 Start the kitchen order calculator
e71d2c2 Add mission m01 guide
```

내가 만든 커밋 4개가 위에 쌓였습니다. 아래 세 개는 미션 브랜치에 원래 있던 커밋입니다.

## 2. 무엇을 어떻게 고칠지 정하기

리뷰 요청을 커밋마다 나눠 봅니다.

| 커밋 | 할 일 |
|---|---|
| `fix typo` | 메시지를 `Fix a typo in the README`로 바꿉니다 |
| `oops` | `fix typo`에 합칩니다. 두 커밋을 합쳐야 오타 수정 하나가 완성됩니다 |
| `Add a note about opening hours` | 그대로 둡니다 |
| `wip` | 버립니다 |

`git rebase -i`는 커밋 목록을 편집기에 띄워 줍니다. 그 목록에서 커밋마다 할 일을 적고 저장하면, git이 위에서부터 한 커밋씩 다시 만듭니다.

## 3. 커밋 목록 열어 보기

### 3-1 git rebase -i

`HEAD~4`는 "지금 커밋에서 4개 전"입니다. 그 위의 커밋 4개, 즉 내가 만든 커밋들을 다시 만들겠다는 뜻입니다.

입력:

```
git rebase -i HEAD~4
```

편집기가 열리고 아래 내용이 나옵니다.

```
pick ??????? # fix typo
pick ??????? # Add a note about opening hours
pick ??????? # oops
pick ??????? # wip

# Rebase 6389ba2..??????? onto 6389ba2 (4 commands)
#
# Commands:
# p, pick <commit> = use commit
# r, reword <commit> = use commit, but edit the commit message
# e, edit <commit> = use commit, but stop for amending
# s, squash <commit> = use commit, but meld into previous commit
# f, fixup [-C | -c] <commit> = like "squash" but keep only the previous
#                    commit's log message, unless -C is used, in which case
#                    keep only this commit's message; -c is same as -C but
#                    opens the editor
# x, exec <command> = run command (the rest of the line) using shell
# b, break = stop here (continue rebase later with 'git rebase --continue')
# d, drop <commit> = remove commit
# l, label <label> = label current HEAD with a name
# t, reset <label> = reset HEAD to a label
# m, merge [-C <commit> | -c <commit>] <label> [# <oneline>]
#         create a merge commit using the original merge commit's
#         message (or the oneline, if no original merge commit was
#         specified); use -c <commit> to reword the commit message
# u, update-ref <ref> = track a placeholder for the <ref> to be updated
#                       to this position in the new commits. The <ref> is
#                       updated at the end of the rebase
#
# These lines can be re-ordered; they are executed from top to bottom.
#
# If you remove a line here THAT COMMIT WILL BE LOST.
#
# However, if you remove everything, the rebase will be aborted.
#
```

- `pick` 줄 하나가 커밋 하나입니다. **맨 위가 가장 오래된 커밋**입니다. `git log`와 순서가 반대입니다.
- 커밋 ID 뒤의 `# fix typo` 같은 부분은 알아보기 쉽게 붙여 준 커밋 메시지입니다. 고쳐도 아무 일도 일어나지 않습니다. 오래된 git은 `#` 없이 메시지만 보여 줍니다.
- `#`으로 시작하는 줄은 설명입니다. git이 읽지 않습니다.

### 3-2 아무것도 바꾸지 않고 닫기

지금은 구경만 합니다. 아무것도 고치지 말고 저장한 뒤 편집기를 닫습니다.

| 편집기 | 저장하고 닫기 |
|---|---|
| nano | `Ctrl+O`, Enter, `Ctrl+X` |
| VS Code | `Ctrl+S`(macOS는 `Cmd+S`)로 저장하고 탭을 닫습니다 |

결과:

```
Successfully rebased and updated refs/heads/m01-messy.
```

모든 줄이 `pick`이고 순서도 그대로면 커밋도 그대로 남습니다.

> 00장에서 편집기를 VS Code로 정했다면 `code --wait`로 등록했는지 확인하세요. `--wait`가 없으면 git이 편집기가 닫히기를 기다리지 않고 바로 끝나 버립니다.

## 4. 순서를 바꾸면 무슨 일이 생기나

`oops`를 `fix typo`에 합치려면 두 줄이 붙어 있어야 합니다. 먼저 순서만 바꿔 보다가 생기는 일을 봅니다.

### 4-1 oops를 맨 위로 올리기

입력:

```
git rebase -i HEAD~4
```

편집기에서 `pick ... oops` 줄을 맨 위로 옮기고 저장합니다. 나머지는 그대로 둡니다.

```
pick ??????? # oops
pick ??????? # fix typo
pick ??????? # Add a note about opening hours
pick ??????? # wip
```

결과:

```
Auto-merging README.md
CONFLICT (content): Merge conflict in README.md
error: could not apply ???????... oops
hint: Resolve all conflicts manually, mark them as resolved with
hint: "git add/rm <conflicted_files>", then run "git rebase --continue".
hint: You can instead skip this commit: run "git rebase --skip".
hint: To abort and get back to the state before "git rebase", run "git rebase --abort".
hint: Disable this message with "git config set advice.mergeConflict false"
Could not apply ???????... # oops
```

`oops`는 `receve`를 `receive`로 바꾼 커밋입니다. 그런데 맨 위로 올리면 아직 `receve`가 없는 파일 위에 다시 만들어야 합니다. 바꿀 줄이 예상과 다르니 git이 멈췄습니다. 이것이 충돌입니다.

### 4-2 지금 상태 보기

입력:

```
git status
```

결과:

```
interactive rebase in progress; onto 6389ba2
Last command done (1 command done):
   pick ??????? # oops
Next commands to do (3 remaining commands):
   pick ??????? # fix typo
   pick ??????? # Add a note about opening hours
  (use "git rebase --edit-todo" to view and edit)
You are currently rebasing branch 'm01-messy' on '6389ba2'.
  (fix conflicts and then run "git rebase --continue")
  (use "git rebase --skip" to skip this patch)
  (use "git rebase --abort" to check out the original branch)

Unmerged paths:
  (use "git restore --staged <file>..." to unstage)
  (use "git add <file>..." to mark resolution)
	both modified:   README.md

no changes added to commit (use "git add" and/or "git commit -a")
```

`interactive rebase in progress`: rebase가 중간에 멈춰 있습니다.

### 4-3 rebase 취소하기

충돌을 풀 수도 있지만, 이번에는 순서를 잘못 정한 것이니 처음으로 돌아갑니다.

입력:

```
git rebase --abort
git log --oneline
```

결과:

```
??????? wip
??????? oops
??????? Add a note about opening hours
??????? fix typo
6389ba2 Add the order total
a3ec238 Start the kitchen order calculator
e71d2c2 Add mission m01 guide
```

rebase를 시작하기 전과 똑같은 커밋 4개로 돌아왔습니다. rebase 도중에 헷갈리면 언제든 `git rebase --abort`로 시작 전으로 돌아갈 수 있습니다.

## 6. 정리를 잘못했다면

`drop`한 커밋이 사실 필요한 것이었다면 어떻게 할까요? rebase가 끝난 뒤에는 `--abort`를 쓸 수 없습니다.
git은 브랜치가 가리키던 커밋을 한동안 기록해 둡니다. 03장에서 본 `git reflog`입니다.

### 6-1 reflog 보기

입력:

```
git reflog -8
```

결과:

```
??????? HEAD@{0}: rebase (finish): returning to refs/heads/m01-messy
??????? HEAD@{1}: rebase (pick): Add a note about opening hours
??????? HEAD@{2}: rebase (fixup): Fix a typo in the README
??????? HEAD@{3}: rebase (reword): Fix a typo in the README
??????? HEAD@{4}: rebase: fast-forward
6389ba2 HEAD@{5}: rebase (start): checkout HEAD~4
??????? HEAD@{6}: rebase (abort): returning to refs/heads/m01-messy
6389ba2 HEAD@{7}: rebase (start): checkout HEAD~4
```

맨 위가 가장 최근입니다. 위에서부터 읽어 내려가다 처음 만나는 `rebase (start)`가 5-1에서 rebase를 시작한 순간입니다. 그 **바로 아래 줄**이 시작 직전, 커밋 4개가 있던 상태입니다. 위 결과에서는 `HEAD@{6}`입니다.

그 줄의 설명이 `wip`가 아니라 `rebase (abort)`인 것은 4-3에서 rebase를 취소하면서 그 커밋으로 돌아왔기 때문입니다. reflog의 설명은 "어떻게 그 커밋에 왔는지"를 적습니다.

### 6-2 그 상태로 돌아가기

입력:

```
git reset --hard "HEAD@{6}"
git log --oneline
```

결과:

```
HEAD is now at ??????? wip
??????? wip
??????? oops
??????? Add a note about opening hours
??????? fix typo
6389ba2 Add the order total
a3ec238 Start the kitchen order calculator
e71d2c2 Add mission m01 guide
```

`wip`까지 커밋 4개가 다시 돌아왔습니다. `notes.txt`도 돌아왔습니다.

- 내 화면의 reflog 번호가 다르면 그 번호를 씁니다. 커밋 ID를 써도 됩니다.
- `HEAD@{6}`처럼 따옴표로 감싸는 이유는 일부 터미널이 `{ }`를 다른 뜻으로 읽기 때문입니다.
- `git reset --hard`는 커밋하지 않은 변경을 지웁니다. 쓰기 전에 `git status`로 남은 변경이 없는지 확인합니다.

## 7. 다음부터는 처음부터 깔끔하게

고칠 커밋이 정해져 있다면, 커밋할 때 미리 표시해 둘 수 있습니다.

### 7-1 새 브랜치에서 오타 고치기

입력:

```
git switch -c m01-fixup origin/mission/m01
```

결과:

```
Switched to a new branch 'm01-fixup'
branch 'm01-fixup' set up to track 'origin/mission/m01'.
```

편집기로 `README.md` 마지막 줄의 `recieve`를 `receve`로 고칩니다. 이번에도 일부러 틀립니다.

입력:

```
git commit -am "Fix a typo in the README"
```

결과:

```
[m01-fixup ???????] Fix a typo in the README
 1 file changed, 1 insertion(+), 1 deletion(-)
```

### 7-2 고친 것을 fixup 커밋으로 남기기

`receve`를 `receive`로 고치고 저장합니다.

`--fixup=HEAD`는 "이 커밋은 HEAD 커밋을 고치는 것"이라고 표시합니다.

입력:

```
git commit -a --fixup=HEAD
git log --oneline -3
```

결과:

```
[m01-fixup ???????] fixup! Fix a typo in the README
 1 file changed, 1 insertion(+), 1 deletion(-)
??????? fixup! Fix a typo in the README
??????? Fix a typo in the README
6389ba2 Add the order total
```

메시지가 `fixup! `과 고칠 커밋의 메시지로 자동으로 만들어졌습니다.

### 7-3 --autosquash로 합치기

입력:

## 11. 메시지와 내용도 미리 표시하기

7절의 `--fixup`은 내용만 고치고 메시지는 그대로 둡니다. git 2.32부터는 메시지를 바꾸는 표시도 할 수 있습니다.

| 명령 | 만들어지는 커밋 | `--autosquash`로 합치면 |
|---|---|---|
| `git commit --fixup=<ID>` | `fixup! 메시지` | 내용만 합칩니다. 메시지는 그대로 |
| `git commit --fixup=reword:<ID>` | `amend! 메시지` (내용 없음) | 메시지만 바꿉니다 |
| `git commit --fixup=amend:<ID>` | `amend! 메시지` | 내용을 합치고 메시지도 바꿉니다 |

### 11-1 메시지만 바꾸는 표시

리뷰가 하나 더 달렸습니다.

> 테스트 커밋 메시지는 무엇을 확인하는지가 드러나게 `Test an order with japchae and mandu`로 바꿔 주세요.

입력:

```
git log --oneline -2
```

결과:

```
??????? Mention the new dishes in the README
??????? Add a test for japchae and mandu
```

`Add a test for japchae and mandu`의 ID를 넣어 입력합니다. `<ID>` 자리에 내 화면의 ID를 씁니다.

```
git commit --fixup=reword:<ID>
```

편집기가 열립니다. 첫 줄 `amend! Add a test for japchae and mandu`는 그대로 두고, 빈 줄 아래의 메시지를 `Test an order with japchae and mandu`로 고친 뒤 저장하고 닫습니다.

```
amend! Add a test for japchae and mandu

Test an order with japchae and mandu
```

결과:

```
[m01-review ???????] amend! Add a test for japchae and mandu
??????? amend! Add a test for japchae and mandu
??????? Mention the new dishes in the README
??????? Add a test for japchae and mandu
```

첫 줄의 `amend!`가 "이 커밋의 메시지로 바꿔 달라"는 표시입니다.

### 11-2 -i 없이 합치기

git 2.44부터는 `--autosquash`를 `-i` 없이 쓸 수 있습니다. 목록을 띄우지 않고 표시대로 바로 정리합니다.

입력:

```
git rebase --autosquash HEAD~3
git log --oneline -5
```

결과:

```
Successfully rebased and updated refs/heads/m01-review.
??????? Mention the new dishes in the README
??????? Test an order with japchae and mandu
??????? Add mandu to the menu
??????? Add allergen notes for japchae
??????? Add japchae to the menu
```

테스트 커밋의 메시지만 바뀌었고, `amend!` 커밋은 사라졌습니다.

> git이 2.44보다 오래되면 오류 없이 `Successfully rebased`만 나오고 아무것도 합쳐지지 않습니다. `git log --oneline -5`에 `amend!` 커밋이 그대로 남아 있으면 이 경우입니다. 그때는 `git rebase -i --autosquash HEAD~3`으로 열고 그대로 저장하고 닫습니다. 버전은 `git --version`으로 봅니다.

### 11-3 -i를 쓸 때마다 자동으로

`git rebase -i`를 열 때마다 `--autosquash`를 붙이기 귀찮다면 설정해 둘 수 있습니다.

```
git config --global rebase.autoSquash true
```

이 설정은 `git rebase -i`에만 적용됩니다. `-i` 없이 정리할 때는 11-2처럼 `--autosquash`를 직접 붙입니다.

> 브랜치 여러 개를 줄줄이 이어서 작업한다면 `git rebase -i --update-refs`가 중간 브랜치들도 함께 옮겨 줍니다. 이 과정에서는 쓰지 않습니다.

## 정리

| 하고 싶은 것 | 방법 |
|---|---|
| 최근 커밋 N개 다듬기 | `git rebase -i HEAD~N` |
| 메시지만 바꾸기 | 목록에서 `reword` |
| 바로 위 커밋에 합치기 (메시지 버림) | `fixup` |
| 바로 위 커밋에 합치기 (메시지 다시 씀) | `squash` |
| 커밋 버리기 | `drop` 또는 줄 지우기 |
| 순서 바꾸기 | 줄을 옮기기. 같은 줄을 고친 커밋끼리는 충돌할 수 있습니다 |
| rebase 도중 그만두기 | `git rebase --abort` |
| rebase가 끝난 뒤 되돌리기 | `git reflog`로 찾고 `git reset --hard "HEAD@{N}"` |
| 나중에 합칠 커밋 미리 표시 | `git commit --fixup=<ID>`, 합칠 때 `git rebase -i --autosquash` |
| 메시지만 바꾸자고 미리 표시 | `git commit --fixup=reword:<ID>` |
| 내용과 메시지를 함께 바꾸자고 미리 표시 | `git commit --fixup=amend:<ID>` |
| 표시한 커밋을 목록 없이 합치기 (git 2.44 이상) | `git rebase --autosquash HEAD~N` |
| 중간 커밋의 내용 고치기 | 목록에서 `edit`, 고친 뒤 `git commit --amend`, `git rebase --continue` |
| 커밋 사이에 새 커밋 넣기 | 목록에 `break`, 멈춘 동안 커밋, `git rebase --continue` |
| 커밋 하나를 둘로 나누기 | `edit`에서 멈춘 뒤 `git reset HEAD^`, 나눠서 커밋, `git rebase --continue` |
| 편집기 저장하고 닫기 (nano) | `Ctrl+O`, Enter, `Ctrl+X` |
