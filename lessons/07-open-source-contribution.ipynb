# 07 오픈소스 방식으로 기여하기: fork에서 PR 보내기

06장에서는 팀원 모두가 팀장의 저장소 **하나에 직접** 브랜치를 push했습니다. 서로 아는 사람끼리, 권한을 주고받을 수 있을 때의 방식입니다.
오픈소스 프로젝트는 다릅니다. 처음 보는 사람에게 push 권한을 줄 수 없습니다. 그래서 기여하는 사람은 각자 **자기 fork**에서 작업하고, 원본 저장소에는 PR로만 보냅니다. 원본에 merge하는 것은 관리자(maintainer)입니다.

이 장에서는 팀이 작은 오픈소스 프로젝트 하나를 직접 운영합니다.

1. 팀장이 Organization과 원본 저장소를 만듭니다.
2. 팀원 모두가 그 저장소를 fork합니다.
3. 각자 근무자 명단 `STAFF.md`에 자기 이름을 올리는 PR을 보냅니다. 모두 파일 끝에 한 줄을 더하므로, 먼저 merge된 PR 다음은 반드시 충돌합니다.
4. 두 번째 라운드에서는 merge 순서를 바꿔 같은 일을 한 번 더 합니다.

**GitHub 계정이 두 개 이상 필요합니다.** 2-4명이 한 팀이 되어 각자 자기 계정으로 합니다.

## 따라 하는 방법

- 회색 상자 **입력**의 명령을 터미널에 그대로 입력합니다. 한 번에 한 줄씩 입력합니다.
- 상자 **결과**는 예시 팀에서 실제로 나온 출력입니다. 예시 팀은 이렇습니다.

  | 역할 | 이름 | 하는 일 |
  |---|---|---|
  | 나 | Kim Minji | 명단에 이름을 올리고, 맡은 자리(`grill`)를 적습니다 |
  | 팀원 | Lee Jun | 같은 일을 합니다. 맡은 자리는 `cold dishes`입니다 |
  | 팀장 | Park Seoyeon | Organization과 원본 저장소를 만들고 PR을 merge합니다 |

- 결과의 `<팀 이름>`은 우리 팀 Organization 이름, `<내 아이디>`는 내 GitHub 아이디, `<팀원 아이디>`는 팀원의 GitHub 아이디로 바꿔 읽습니다.
- 이 장의 커밋은 모두 새로 만드는 것이라 ID를 `???????`로 가렸습니다. 내 화면에는 다른 ID가 나옵니다.

## 0. 팀 준비

### 0-1 팀과 팀장 정하기

2-4명이 한 팀입니다. 그중 한 사람이 팀장을 맡습니다. 팀장은 원본 저장소의 관리자가 됩니다.
06장과 달리 팀원은 원본 저장소에 push할 권한이 없습니다. 권한이 없어도 기여할 수 있는 것이 이 방식의 핵심입니다.

### 0-2 팀장: Organization 만들기

팀장만 합니다. Organization은 개인 계정이 아닌 "팀"이 저장소를 가지는 GitHub 계정입니다. 오픈소스 프로젝트 대부분이 Organization 아래에 있습니다.

1. GitHub 오른쪽 위 프로필 사진 → **Your organizations** → **New organization**을 누릅니다.
2. 요금제는 **Free**를 고릅니다.
3. Organization 이름을 정합니다. 이 이름이 주소가 됩니다(`https://github.com/<팀 이름>`).
4. 연락용 이메일을 넣고, 이 Organization이 **My personal account**에 속한다고 고른 뒤 만듭니다.
5. 팀원을 초대하는 화면이 나오면 팀원의 GitHub 아이디를 넣습니다. 건너뛰었다면 Organization의 **People** → **Invite member**에서 초대합니다.

팀원은 초대 메일에서 **Join**을 누릅니다. 초대받은 팀원도 원본 저장소에 push할 권한은 없습니다. 같은 팀으로 보이고, PR 리뷰를 부탁받을 수 있게 하는 것뿐입니다.

### 0-3 팀장: 원본 저장소 만들기

1. Organization 페이지에서 **Repositories** → **New repository**를 누릅니다.
2. **Owner**가 `<팀 이름>`인지 확인합니다. 내 개인 계정으로 되어 있으면 바꿉니다.
3. 이름은 `kitchen-staff`, 공개 범위는 **Public**으로 합니다.
4. **README, .gitignore, license는 모두 추가하지 않고** 빈 저장소로 만듭니다. 첫 내용은 0-4에서 터미널로 올립니다.

### 0-4 팀장: 첫 내용 올리기

팀장의 컴퓨터에서 합니다. 이 저장소는 처음부터 새로 만드는 것이라 `clone`이 아니라 `init`으로 시작합니다.

입력:

```
cd ~
mkdir kitchen-staff
cd kitchen-staff
git init -b main
```

결과:

```
Initialized empty Git repository in ~/kitchen-staff/.git/
```

`-b main`은 첫 브랜치 이름을 `main`으로 정합니다.

편집기로 `STAFF.md`를 만들고 이렇게 씁니다. 마지막 줄은 팀장 자신입니다.

```
# Kitchen staff

One line per person. Add your station after a colon when you know it.

- Park Seoyeon: front of house
```

입력:

```
git add STAFF.md
git commit -m "Start the staff list"
git remote add origin https://github.com/<팀 이름>/kitchen-staff.git
git push -u origin main
```

결과:

```
[main (root-commit) ???????] Start the staff list
 1 file changed, 5 insertions(+)
 create mode 100644 STAFF.md
To https://github.com/<팀 이름>/kitchen-staff.git
 * [new branch]      main -> main
branch 'main' set up to track 'origin/main'.
```

팀장이 원본에 직접 push하는 것은 이 한 번뿐입니다. 이제부터는 팀장도 다른 사람의 PR을 merge만 합니다.

### 0-5 모두: fork 만들기

팀장을 포함해 모두가 합니다.

1. `https://github.com/<팀 이름>/kitchen-staff`를 엽니다.
2. 오른쪽 위 **Fork**를 누릅니다.
3. **Owner**를 내 개인 계정으로 고르고 **Create fork**를 누릅니다.

`https://github.com/<내 아이디>/kitchen-staff`가 생깁니다. 00장에서 `git-practice`를 fork한 것과 같은 일입니다.

## 1. 내 fork 받기

### 1-1 clone

입력:

```
cd ~
git clone https://github.com/<내 아이디>/kitchen-staff.git
```

결과:

```
Cloning into 'kitchen-staff'...
done.
```

### 1-2 원본을 upstream으로 등록

`origin`은 내 fork입니다. 원본 저장소에도 이름을 붙여 둡니다. 05장에서 한 것과 같습니다.

입력:

```
cd kitchen-staff
git remote add upstream https://github.com/<팀 이름>/kitchen-staff.git
git remote -v
```

결과:

```
origin	https://github.com/<내 아이디>/kitchen-staff.git (fetch)
origin	https://github.com/<내 아이디>/kitchen-staff.git (push)
upstream	https://github.com/<팀 이름>/kitchen-staff.git (fetch)
upstream	https://github.com/<팀 이름>/kitchen-staff.git (push)
```

### 1-3 원본 받아 오기

입력:

```
git fetch upstream
```

결과:

```
From https://github.com/<팀 이름>/kitchen-staff
 * [new branch]      main       -> upstream/main
```

### 퀴즈 07-1

`git push origin`과 `git push upstream` 중 내가 할 수 있는 것은 어느 쪽입니까? 다른 쪽을 하면 GitHub는 어떻게 답할까요? 팀장이라면 답이 달라집니까?

## 2. 내 이름을 올리는 커밋

### 2-1 브랜치 만들기

fork 안에서도 `main`에 바로 커밋하지 않습니다. 이유는 6절에서 봅니다.

입력:

```
git switch -c add-minji
```

결과:

```
Switched to a new branch 'add-minji'
```

### 2-2 명단에 한 줄 추가

편집기로 `STAFF.md`를 열고 맨 끝에 내 이름을 한 줄 추가합니다. 맡은 자리는 아직 쓰지 않습니다.

```
- Park Seoyeon: front of house
- Kim Minji
```

입력:

```
git diff
```

결과:

```
diff --git a/STAFF.md b/STAFF.md
index aac8380..ac31694 100644
--- a/STAFF.md
+++ b/STAFF.md
@@ -3,3 +3,4 @@
 One line per person. Add your station after a colon when you know it.
 
 - Park Seoyeon: front of house
+- Kim Minji
```

### 2-3 서명을 붙여 커밋하기

이 팀은 "커밋마다 `Signed-off-by` 줄을 붙인다"는 규칙을 정했다고 합시다. Linux 커널처럼 이 줄을 요구하는 오픈소스 프로젝트가 있습니다. "이 변경을 이 프로젝트의 라이선스로 내놓을 권리가 나에게 있다"는 확인입니다. `-s`를 붙이면 git이 내 이름과 이메일로 이 줄을 써 줍니다.

입력:

```
git add STAFF.md
git commit -s -m "Add Kim Minji to the staff list"
git log -1
```

결과:

```
[add-minji ???????] Add Kim Minji to the staff list
 1 file changed, 1 insertion(+)
commit ????????????????????????????????????????
Author: Kim Minji <minji@example.invalid>
Date:   Fri Nov 20 10:00:00 2026 +0900

    Add Kim Minji to the staff list
    
    Signed-off-by: Kim Minji <minji@example.invalid>
```

메시지 맨 끝에 `Signed-off-by: Kim Minji <minji@example.invalid>`가 붙었습니다. 이 줄은 글자일 뿐이고, 누가 썼는지 암호로 증명하는 "서명된 커밋"(GitHub의 **Verified** 표시)과는 다른 것입니다.

> 서명을 요구하지 않는 프로젝트라면 `-s`는 붙이지 않아도 됩니다. 기여하기 전에 그 프로젝트의 `CONTRIBUTING.md`를 먼저 읽고 규칙을 따릅니다.

### 2-4 내 fork에 push

입력:

```
git push -u origin add-minji
```

결과:

```
To https://github.com/<내 아이디>/kitchen-staff.git
 * [new branch]      add-minji -> add-minji
branch 'add-minji' set up to track 'origin/add-minji'.
```

## 3. 원본 저장소로 PR 보내기

### 3-1 PR 만들기

1. `https://github.com/<내 아이디>/kitchen-staff`를 엽니다.
2. 노란 띠의 **Compare & pull request**를 누릅니다. 띠가 없으면 **Contribute** → **Open pull request**를 누릅니다.
3. 화면 위쪽의 네 칸을 확인합니다.

| 칸 | 값 |
|---|---|
| base repository | `<팀 이름>/kitchen-staff` (원본) |
| base | `main` |
| head repository | `<내 아이디>/kitchen-staff` (내 fork) |
| compare | `add-minji` |

4. 아래쪽 **Allow edits by maintainers** 체크를 켜 둡니다.
5. 제목과 설명을 쓰고 **Create pull request**를 누릅니다.

> **base repository가 내 fork(`<내 아이디>/kitchen-staff`)로 되어 있으면** PR이 내 저장소 안에서만 열립니다. 팀장은 볼 수 없습니다. 06장과 반대로, 이번에는 base가 **원본**이어야 맞습니다.

**Allow edits by maintainers**를 켜면 원본의 관리자가 내 fork의 `add-minji` 브랜치에 직접 커밋을 push할 수 있습니다. 오타 하나 때문에 리뷰를 한 번 더 주고받지 않고 관리자가 바로 고쳐 줄 수 있습니다.

### 퀴즈 07-2

관리자가 **Allow edits by maintainers** 덕분에 내 `add-minji`에 오타 수정 커밋을 하나 push했습니다. 그 사실을 모르고 내가 `git commit --amend`를 한 뒤 `git push --force`를 하면 어떻게 됩니까? `--force-with-lease`였다면 어떻게 됩니까?

## 4. 팀원의 PR을 내 컴퓨터에서 보기

예시 팀에서는 Jun이 Minji보다 먼저 PR을 열었습니다. 그래서 Jun의 PR이 1번, Minji의 PR이 2번입니다.

### 4-1 PR 번호로 받아 오기

리뷰하는 사람은 PR의 코드를 자기 컴퓨터에서 돌려 보고 싶을 때가 많습니다. 그런데 Jun의 브랜치는 Jun의 fork에 있습니다. 그 fork를 원격으로 등록하지 않아도, GitHub는 원본 저장소에 PR마다 `pull/<번호>/head`라는 이름을 만들어 둡니다.

입력:

```
git fetch upstream pull/1/head:pr-1
```

결과:

```
From https://github.com/<팀 이름>/kitchen-staff
 * [new ref]         refs/pull/1/head -> pr-1
```

`pull/1/head:pr-1`은 "원본의 `pull/1/head`를 받아서 내 컴퓨터에 `pr-1`이라는 브랜치로 만든다"는 뜻입니다.

### 4-2 받은 PR 읽기

입력:

```
git show pr-1
```

결과:

```
commit ????????????????????????????????????????
Author: Lee Jun <jun@example.invalid>
Date:   Fri Nov 20 10:00:00 2026 +0900

    Add Lee Jun to the staff list
    
    Signed-off-by: Lee Jun <jun@example.invalid>

diff --git a/STAFF.md b/STAFF.md
index aac8380..4b9be42 100644
--- a/STAFF.md
+++ b/STAFF.md
@@ -3,3 +3,4 @@
 One line per person. Add your station after a colon when you know it.
 
 - Park Seoyeon: front of house
+- Lee Jun
```

브랜치로 받았으니 `git switch pr-1`로 옮겨서 테스트를 돌려 볼 수도 있습니다.

> GitHub CLI(`gh`)를 쓴다면 `gh pr checkout 1` 한 줄로 받아서 바로 옮겨 갑니다. 00장에서 `gh`에 로그인했다면 이 폴더에서 그대로 쓸 수 있습니다.

### 퀴즈 07-3

리뷰어로서 Jun의 PR에서 무엇을 확인하겠습니까? 이 팀의 규칙(파일 끝에 한 줄, `Signed-off-by` 줄)을 기준으로 4-2의 결과에서 두 가지를 찾아 적어 보세요.

## 5. 다른 사람의 PR이 먼저 merge됐을 때

### 5-1 GitHub에서 보이는 것

팀장이 Jun의 PR을 먼저 merge했습니다. 그러자 Minji의 PR 아래쪽에 **This branch has conflicts that must be resolved**가 나오고 merge 버튼을 누를 수 없게 되었습니다.
두 사람 모두 `STAFF.md`의 같은 자리, 파일 끝에 줄을 더했기 때문입니다.

오픈소스 프로젝트에서는 이럴 때 관리자가 "rebase 해 주세요"라고 댓글을 답니다. 충돌을 푸는 것은 PR을 보낸 사람의 일입니다.

### 5-2 원본의 최신 상태 받기

내 fork가 아니라 **원본**에서 받습니다. 새 커밋은 원본에 들어갔기 때문입니다.

입력:

```
git fetch upstream
git log --oneline add-minji..upstream/main
```

결과:

```
From https://github.com/<팀 이름>/kitchen-staff
   ???????..???????  main       -> upstream/main
??????? Merge pull request #1 from <팀원 아이디>/add-jun
??????? Add Lee Jun to the staff list
```

### 5-3 원본의 main 위로 옮기기

입력:

```
git rebase upstream/main
```

결과:

```
Auto-merging STAFF.md
CONFLICT (content): Merge conflict in STAFF.md
error: could not apply ???????... Add Kim Minji to the staff list
hint: Resolve all conflicts manually, mark them as resolved with
hint: "git add/rm <conflicted_files>", then run "git rebase --continue".
hint: You can instead skip this commit: run "git rebase --skip".
hint: To abort and get back to the state before "git rebase", run "git rebase --abort".
hint: Disable this message with "git config set advice.mergeConflict false"
Could not apply ???????... # Add Kim Minji to the staff list
```

### 5-4 충돌 난 파일 확인

입력:

```
cat STAFF.md
```

결과:

```
# Kitchen staff

One line per person. Add your station after a colon when you know it.

- Park Seoyeon: front of house
<<<<<<< HEAD
- Lee Jun
||||||| parent of ??????? (Add Kim Minji to the staff list)
=======
- Kim Minji
>>>>>>> ??????? (Add Kim Minji to the staff list)
```

위쪽(`HEAD`)이 원본에 이미 들어간 Jun의 줄, 아래쪽이 옮기는 중인 내 줄입니다. 가운데 `|||||||` 부분은 두 사람이 고치기 전의 원래 줄(05장 8절)인데, 둘 다 새 줄을 더한 것이라 비어 있습니다.

### 5-5 두 줄 모두 살리기

표시 줄을 모두 지우고 두 이름을 모두 남깁니다. 먼저 들어간 Jun을 위에 둡니다.

```
- Park Seoyeon: front of house
- Lee Jun
- Kim Minji
```

입력:

```
git add STAFF.md
git rebase --continue
```

결과:

```
[detached HEAD ???????] Add Kim Minji to the staff list
 1 file changed, 1 insertion(+)
Successfully rebased and updated refs/heads/add-minji.
```

### 5-6 PR의 이전 판과 새 판 비교하기

push하기 전에, 고친 커밋이 원래 커밋과 무엇이 다른지 확인합니다. 아직 push하지 않았으니 `origin/add-minji`는 GitHub에 올라가 있는 **이전 판**입니다.

입력:

```
git range-diff --creation-factor=100 upstream/main origin/add-minji add-minji
```

결과:

```
1:  ??????? ! 1:  ??????? Add Kim Minji to the staff list
    @@ Commit message
         Signed-off-by: Kim Minji <minji@example.invalid>
     
      ## STAFF.md ##
    -@@
    - One line per person. Add your station after a colon when you know it.
    +@@ STAFF.md: One line per person. Add your station after a colon when you know it.
      
      - Park Seoyeon: front of house
    + - Lee Jun
     +- Kim Minji
```

`range-diff`는 커밋 묶음 두 개를 커밋끼리 짝지어 비교합니다. `--creation-factor=100`은 "많이 달라 보여도 짝을 지어 보라"는 옵션입니다. 이번처럼 한 줄짜리 커밋은 주변 줄만 바뀌어도 git이 다른 커밋으로 보고 짝을 짓지 않기 때문입니다.

- 첫 줄 `1: ??????? ! 1: ???????`: 이전 판 1번 커밋과 새 판 1번 커밋이 짝이고, 내용이 달라졌다(`!`)는 뜻입니다. 같으면 `=`가 나옵니다.
- 그 아래는 **변경 내용을 한 번 더 diff한 것**입니다. 맨 앞 칸의 `-`는 이전 판에만, `+`는 새 판에만 있는 줄입니다. 둘째 칸의 기호는 원래 커밋의 변경입니다.
- `+ - Lee Jun`: 새 판에서는 내 줄 바로 위에 Jun의 줄이 보입니다. `@@` 줄이 바뀐 것은 위치 정보가 달라졌을 뿐입니다.
- 내 커밋이 더한 줄 ` +- Kim Minji`와 메시지의 `Signed-off-by` 줄은 맨 앞 칸이 비어 있습니다. 그대로라는 뜻입니다.

리뷰어도 같은 것을 봅니다. 덮어쓰는 push가 들어오면 GitHub PR 화면에 `force-pushed` 줄과 **Compare** 링크가 생기고, 그 링크가 이전 판과 새 판의 차이를 보여 줍니다.

### 5-7 다시 push

입력:

```
git push --force-with-lease
```

결과:

```
To https://github.com/<내 아이디>/kitchen-staff.git
 + ???????...??????? add-minji -> add-minji (forced update)
```

내 fork의 `add-minji`가 바뀌면 PR도 저절로 새 판으로 바뀝니다. PR을 새로 열지 않습니다.

## 6. merge된 뒤: 내 fork의 main 맞추기

팀장이 Minji의 PR을 merge했습니다. 원본의 `main`에는 두 이름이 다 들어갔습니다. 그런데 **내 fork의 `main`은 그대로**입니다. fork는 만든 순간의 복사본이라, 원본이 바뀌어도 저절로 따라가지 않습니다.

### 6-1 main으로 이동

입력:

```
git switch main
```

결과:

```
Switched to branch 'main'
Your branch is up to date with 'origin/main'.
```

`up to date`라고 나오지만 믿으면 안 됩니다. 비교한 상대가 `origin/main`, 즉 **내 fork**이기 때문입니다. 내 fork도 원본보다 뒤처져 있습니다.

### 6-2 원본을 받아서 맞추기

입력:

```
git fetch upstream
git merge --ff-only upstream/main
```

결과:

```
From https://github.com/<팀 이름>/kitchen-staff
   ???????..???????  main       -> upstream/main
Updating ???????..???????
Fast-forward
 STAFF.md | 2 ++
 1 file changed, 2 insertions(+)
```

`--ff-only`는 "앞으로 당기기만 하고, 그럴 수 없으면 멈춘다"는 뜻입니다. 내 `main`에 원본에 없는 커밋이 있으면 당기지 못하고 멈춥니다. 2-1에서 `main`에 바로 커밋하지 않은 이유가 이것입니다. `main`을 깨끗하게 두면 언제든 이 두 줄로 원본과 똑같이 맞출 수 있습니다.

### 6-3 내 fork에도 올리기

입력:

```
git push origin main
```

결과:

```
To https://github.com/<내 아이디>/kitchen-staff.git
   ???????..???????  main -> main
```

이제 내 컴퓨터, 내 fork, 원본의 `main`이 모두 같습니다.

> GitHub 화면에서 해도 됩니다. 내 fork 페이지의 **Sync fork** → **Update branch**를 누르면 원본의 `main`을 내 fork에 가져옵니다. `gh`를 쓴다면 `gh repo sync <내 아이디>/kitchen-staff`도 같은 일을 합니다. 어느 쪽이든 그다음 내 컴퓨터에서는 `git pull`로 받습니다.

### 6-4 다 쓴 브랜치 지우기

입력:

```
git branch -d add-minji pr-1
git push origin --delete add-minji
```

결과:

```
Deleted branch add-minji (was ???????).
Deleted branch pr-1 (was ???????).
To https://github.com/<내 아이디>/kitchen-staff.git
 - [deleted]         add-minji
```

## 7. 두 번째 라운드: merge 순서 바꾸기

이번에는 명단의 내 이름 옆에 맡은 자리를 적습니다. 이번에는 **내 PR이 먼저** merge되고, 팀원이 충돌을 풉니다.

### 7-1 브랜치, 수정, 커밋, push

방금 맞춘 `main`에서 새 브랜치를 만듭니다. 편집기로 내 줄을 `- Kim Minji: grill`로 고칩니다.

입력:

```
git switch -c station-minji
git diff
```

결과:

```
Switched to a new branch 'station-minji'
diff --git a/STAFF.md b/STAFF.md
index 1b2586d..4d0d449 100644
--- a/STAFF.md
+++ b/STAFF.md
@@ -4,4 +4,4 @@ One line per person. Add your station after a colon when you know it.
 
 - Park Seoyeon: front of house
 - Lee Jun
-- Kim Minji
+- Kim Minji: grill
```

입력:

```
git commit -s -am "Add Kim Minji's station"
git push -u origin station-minji
```

결과:

```
[station-minji ???????] Add Kim Minji's station
 1 file changed, 1 insertion(+), 1 deletion(-)
To https://github.com/<내 아이디>/kitchen-staff.git
 * [new branch]      station-minji -> station-minji
branch 'station-minji' set up to track 'origin/station-minji'.
```

`-a`는 이미 git이 알고 있는 파일의 변경을 `git add` 없이 바로 커밋에 넣습니다. 새로 만든 파일은 넣지 않습니다.

3-1처럼 원본으로 PR을 엽니다. 예시 팀에서는 이 PR이 3번, 같은 때 Jun이 연 PR이 4번입니다.

### 7-2 이번에는 팀원 쪽이 충돌

팀장이 Minji의 PR을 먼저 merge했습니다. 이번에는 Jun의 PR에 충돌 표시가 나옵니다. Jun과 Minji의 줄이 바로 붙어 있어서, git은 두 변경이 같은 자리라고 봅니다.

이번에 Minji가 할 일은 리뷰어 역할입니다. Jun의 PR에 이렇게 댓글을 남깁니다.

> 3번 PR이 먼저 merge되어 충돌이 났습니다. `upstream/main` 위로 rebase해 주세요.

Jun은 5절에서 Minji가 한 것을 그대로 합니다. `git fetch upstream`, `git rebase upstream/main`, 충돌 풀기, `git push --force-with-lease`.

### 7-3 모두 merge된 뒤 맞추기

Jun이 rebase한 PR까지 merge되었습니다. 6절과 같은 순서로 맞춥니다. 이번에는 받기와 당기기를 한 줄로 합니다.

입력:

```
git switch main
git pull --ff-only upstream main
git push origin main
```

결과:

```
Switched to branch 'main'
Your branch is up to date with 'origin/main'.
From https://github.com/<팀 이름>/kitchen-staff
 * branch            main       -> FETCH_HEAD
   ???????..???????  main       -> upstream/main
Updating ???????..???????
Fast-forward
 STAFF.md | 4 ++--
 1 file changed, 2 insertions(+), 2 deletions(-)
To https://github.com/<내 아이디>/kitchen-staff.git
   ???????..???????  main -> main
```

`git pull --ff-only upstream main`은 6-2의 `git fetch upstream`과 `git merge --ff-only upstream/main`을 한 번에 합니다.

## 8. 실제 오픈소스 PR 읽어 보기

이제 다른 프로젝트에서 이 과정이 어떻게 오가는지 직접 찾아 봅니다. 프로젝트는 내가 고릅니다. 내가 쓰는 라이브러리나 도구, 또는 관심 있는 분야의 프로젝트면 됩니다. 별(★)이 많고 최근까지 PR이 merge되는 프로젝트가 좋습니다.

1. 그 프로젝트의 **Pull requests** 탭을 엽니다.
2. 검색 칸에 `is:pr is:merged review:changes_requested`를 넣습니다. "수정 요청을 받은 뒤 merge된 PR"만 나옵니다.
3. 하나를 골라 열고, 아래 표를 채웁니다.

| 찾을 것 | 어디서 보나 |
|---|---|
| 리뷰어가 고쳐 달라고 한 것 | **Conversation** 탭의 리뷰 댓글 |
| PR을 보낸 사람이 어떻게 고쳤나: 커밋을 더했나, 덮어썼나 | **Conversation** 탭의 `force-pushed` 줄과 **Commits** 탭 |
| 커밋 메시지의 본문: 제목 아래에 무엇을 썼나 | **Commits** 탭에서 커밋을 열기 |
| `Signed-off-by` 같은 끝줄이 있나 | 같은 커밋 화면의 메시지 끝 |
| 기여 규칙 문서 | 저장소의 `CONTRIBUTING.md` |

## 정리

### 06장의 방식과 07장의 방식

| | 06 팀 저장소 (Collaborator) | 07 fork에서 PR |
|---|---|---|
| 내 브랜치를 push하는 곳 | 팀 저장소 (`origin`) | 내 fork (`origin`) |
| 원본에 push 권한 | 있음 | 없음 |
| 최신 변경을 받는 곳 | `origin` | `upstream` |
| PR 방향 | 팀 저장소 안에서 브랜치 → `main` | 내 fork의 브랜치 → 원본의 `main` |
| 주로 쓰는 곳 | 회사 팀, 서로 아는 소수 | 오픈소스, 처음 보는 기여자 |

### 이 장에서 새로 쓴 명령

| 상황 | 명령 |
|---|---|
| 새 저장소 시작 | `git init -b main` |
| 원본 등록 | `git remote add upstream 주소` |
| 커밋에 Signed-off-by 붙이기 | `git commit -s` |
| 남의 PR을 브랜치로 받기 | `git fetch upstream pull/번호/head:pr-번호` 또는 `gh pr checkout 번호` |
| PR의 이전 판과 새 판 비교 | `git range-diff upstream/main origin/브랜치 브랜치` |
| 내 fork의 main을 원본에 맞추기 | `git switch main`, `git pull --ff-only upstream main`, `git push origin main` |

### 오픈소스에 처음 기여할 때

- [ ] `CONTRIBUTING.md`를 읽었다 (서명, 커밋 메시지, 테스트 규칙)
- [ ] fork를 만들고, 원본을 `upstream`으로 등록했다
- [ ] 작업을 시작하기 전에 내 `main`을 원본에 맞췄다
- [ ] `main`이 아니라 새 브랜치에서 작업했다
- [ ] PR의 base repository가 원본인지 확인했다
- [ ] 리뷰에서 rebase를 부탁받으면 `upstream/main` 위로 rebase하고 `--force-with-lease`로 올린다
- [ ] merge된 뒤 내 fork의 `main`을 맞추고 브랜치를 지웠다
