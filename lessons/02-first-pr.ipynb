# 02 내 첫 변경을 올리고 Pull Request 열기

kitchen 프로젝트의 README를 읽다가 오타를 하나 발견했습니다. `recieve`가 아니라 `receive`가 맞습니다.
한 글자 고치는 일이지만, 팀에서는 이 수정도 정해진 길을 거쳐 들어갑니다.

1. 내 브랜치를 만들어 그 위에서 고칩니다.
2. 바뀐 내용을 확인하고 커밋으로 남깁니다.
3. 내 fork에 push합니다.
4. 원래 저장소에 Pull Request를 열어 "이 변경을 받아 주세요"라고 요청합니다.

이 장에서 이 길을 한 번 끝까지 걷습니다. 미션 m01이 이 장의 실습입니다.
9절부터는 실제 프로젝트에 버그 수정을 보낼 때 커밋에 무엇을 더 남기는지 연습합니다. 본문과 이슈 번호, 커밋 나누기, Signed-off-by입니다.

## 따라 하는 방법

- 회색 상자 **입력**의 명령을 터미널에 그대로 입력하고 Enter를 누릅니다. 한 번에 한 줄씩 입력합니다.
- 상자 **결과**는 이 저장소에서 실제로 나오는 출력입니다. 내 화면과 비교해 보세요.
- 내가 만든 커밋의 ID는 `???????`로 가렸습니다. 내 이름과 시각이 커밋에 들어가기 때문에, 내 화면에는 사람마다 다른 ID가 나옵니다. 원래 있던 커밋의 ID는 똑같이 나와야 정상입니다.
- 결과에 나오는 이름 `Kim Minji`와 `<내 아이디>` 자리에는 내 이름과 내 GitHub 아이디가 나옵니다.
- **퀴즈**는 답을 싣지 않았습니다. 직접 해 보고 적어 두세요.

## 1. 내 브랜치 만들기

`mission/m01`에서 바로 고쳐도 커밋은 됩니다. 그런데 팀에서는 보통 그렇게 하지 않습니다.

- 다른 사람과 같이 쓰는 브랜치를 내 작업으로 바로 바꾸면, 아직 검토받지 않은 변경이 섞입니다.
- 작업 하나에 브랜치 하나를 두면, 잘못됐을 때 그 브랜치만 버리면 됩니다.
- Pull Request는 "이 브랜치의 변경을 저 브랜치에 넣어 주세요"라는 요청이라, 내 브랜치가 따로 있어야 합니다.

### 1-1 새 브랜치를 만들고 이동하기

`-c`는 create, 새로 만든다는 뜻입니다. 브랜치 이름은 무엇을 하는지 알 수 있게 짓습니다.

입력:

```
git switch -c m01-typo
```

결과:

```
Switched to a new branch 'm01-typo'
```

### 1-2 지금 어느 브랜치에 있는지 보기

입력:

```
git branch
```

결과:

```
* m01-typo
  main
  mission/m01
```

`*`가 붙은 것이 지금 있는 브랜치입니다. `m01-typo`는 `mission/m01`에서 갈라져 나왔으므로, 지금은 두 브랜치의 내용이 같습니다.

## 2. 고치고 확인하기

### 2-1 README 고치기

편집기로 `README.md`를 엽니다. 마지막 줄의 `recieve`를 `receive`로 고치고 저장합니다.

### 2-2 무엇이 바뀌었는지 보기: git status

입력:

```
git status
```

결과:

```
On branch m01-typo
Changes not staged for commit:
  (use "git add <file>..." to update what will be committed)
  (use "git restore <file>..." to discard changes in working directory)
	modified:   README.md

no changes added to commit (use "git add" and/or "git commit -a")
```

`Changes not staged for commit` 아래에 `README.md`가 있습니다. "파일은 바뀌었지만, 아직 다음 커밋에 넣겠다고 정하지 않았다"는 뜻입니다.

### 2-3 어떻게 바뀌었는지 보기: git diff

입력:

```
git diff
```

결과:

```
diff --git a/README.md b/README.md
index e73ac76..5959517 100644
--- a/README.md
+++ b/README.md
@@ -7,4 +7,4 @@ python3 -m unittest
 ```
 
 When a customer orders, the cashier enters each dish and its quantity,
-and the total is shown before they recieve the bill.
+and the total is shown before they receive the bill.
```

| 줄 | 뜻 |
|---|---|
| `-`로 시작 | 지우는 줄 |
| `+`로 시작 | 새로 들어가는 줄 |
| 앞에 공백 | 바뀌지 않은 앞뒤 줄. 어디쯤인지 알려 주려고 함께 보입니다 |
| `@@ -7,4 +7,4 @@` | 7번째 줄부터 4줄을 보여 준다는 표시 |

**커밋하기 전에 diff를 보는 습관**을 들이세요. 의도하지 않은 변경(편집기가 바꾼 공백, 실수로 지운 줄)은 여기서 잡힙니다.

## 3. 다음 커밋에 넣을 파일 고르기

git은 바뀐 파일을 모두 커밋하지 않습니다. 이번 커밋에 넣을 변경을 먼저 고르고, 고른 것만 커밋합니다.
고른 변경이 모이는 곳을 **스테이징 영역**(staging area)이라고 부릅니다. 커밋 직전의 대기석입니다.

### 3-1 git add

입력:

```
git add README.md
```

성공하면 아무것도 나오지 않습니다.


### 3-2 다시 git status

입력:

```
git status
```

결과:

```
On branch m01-typo
Changes to be committed:
  (use "git restore --staged <file>..." to unstage)
	modified:   README.md
```

`README.md`가 `Changes to be committed`로 옮겨 갔습니다. 이제 커밋하면 이 변경이 들어갑니다.
괄호 안의 `git restore --staged`는 고른 것을 취소하는 명령입니다. 03장에서 배웁니다.

### 3-3 고른 변경 보기: git diff --staged

`git add`를 한 뒤에는 `git diff`에 아무것도 나오지 않습니다. 고른 변경은 `--staged`를 붙여야 보입니다.

입력:

```
git diff --staged
```

결과:

```
diff --git a/README.md b/README.md
index e73ac76..5959517 100644
--- a/README.md
+++ b/README.md
@@ -7,4 +7,4 @@ python3 -m unittest
 ```
 
 When a customer orders, the cashier enters each dish and its quantity,
-and the total is shown before they recieve the bill.
+and the total is shown before they receive the bill.
```

| 명령 | 보여 주는 것 |
|---|---|
| `git diff` | 바뀌었지만 아직 고르지 않은 변경 |
| `git diff --staged` | 골라서 다음 커밋에 들어갈 변경 |

확인한 뒤에는 추가한 빈 줄을 지우고 저장해서 3-3 상태로 돌려놓습니다.

## 4. 커밋하기

### 4-1 git commit -m

`-m` 뒤에 커밋 메시지를 씁니다.

입력:

```
git commit -m "Fix a typo in the README"
```

결과:

```
[m01-typo ???????] Fix a typo in the README
 1 file changed, 1 insertion(+), 1 deletion(-)
```

| 부분 | 뜻 |
|---|---|
| `m01-typo` | 커밋이 들어간 브랜치 |
| `???????` | 새 커밋의 ID 앞부분 (내 화면에는 실제 ID) |
| `1 file changed, 1 insertion(+), 1 deletion(-)` | 파일 1개, 한 줄을 지우고 한 줄을 넣었다는 뜻 |

### 4-2 좋은 커밋 메시지

01장에서 이력을 읽을 때 커밋 메시지만 보고 무엇이 바뀌었는지 알 수 있었습니다. 그게 가능했던 건 메시지를 이렇게 썼기 때문입니다.

| 규칙 | 좋은 예 | 나쁜 예 |
|---|---|---|
| 첫 줄은 72자 이내 | `Fix a typo in the README` | 한 줄에 바꾼 내용을 다 나열 |
| 무엇을 했는지 동사로 시작 | `Fix ...`, `Add ...`, `Change ...` | `fixed`, `수정함` |
| 커밋 하나에 변경 하나 | 오타 수정만 | `Fix typo and add kimbap` |
| 남이 읽어도 알 수 있게 | `Round each share to cents` | `fix`, `asdf`, `final2` |

영어로 쓰는 팀이 많지만, 팀이 한국어로 쓰기로 했다면 한국어도 괜찮습니다. 중요한 것은 팀 안에서 같은 방식을 쓰는 것입니다.

## 7. 내 fork에 올리기

지금까지의 커밋은 내 컴퓨터에만 있습니다. GitHub에 있는 내 fork로 보내야 다른 사람이 볼 수 있습니다. 이것을 push라고 합니다.

### 7-1 git push -u

입력:

```
git push -u origin m01-typo
```

결과:

```
To https://github.com/<내 아이디>/git-practice.git
 * [new branch]      m01-typo -> m01-typo
branch 'm01-typo' set up to track 'origin/m01-typo'.
```

| 부분 | 뜻 |
|---|---|
| `origin` | 보낼 곳. clone해 온 내 fork입니다 (00장 6-1에서 확인했습니다) |
| `m01-typo` | 보낼 브랜치 |
| `-u` | 내 브랜치와 fork의 브랜치를 짝지어 둡니다. 다음부터는 `git push`만 입력해도 같은 곳으로 갑니다 |
| `* [new branch]` | fork에 같은 이름의 브랜치가 새로 생겼다는 뜻 |

GitHub에 push하면 결과 위쪽에 `remote:`로 시작하는 줄이 몇 줄 더 나옵니다. 그 안에 Pull Request를 바로 열 수 있는 주소가 있습니다.

### 7-2 짝이 맞는지 보기

입력:

```
git status
```

결과:

```
On branch m01-typo
Your branch is up to date with 'origin/m01-typo'.

nothing to commit, working tree clean
```

`up to date with 'origin/m01-typo'`는 "내 브랜치와 fork의 브랜치가 같은 커밋을 가리킨다"는 뜻입니다.

## 8. Pull Request 열기

이제 원래 저장소인 `Drillcheck/git-practice`에 "내 `m01-typo`의 변경을 `mission/m01`에 받아 주세요"라고 요청합니다. 이 요청이 **Pull Request**(PR)입니다.

### 8-1 PR 화면 열기

1. 브라우저에서 내 fork `https://github.com/<내 아이디>/git-practice`를 엽니다.
2. 위쪽에 노란 띠로 **m01-typo had recent pushes**와 **Compare & pull request** 버튼이 보입니다. 버튼을 누릅니다.
   띠가 보이지 않으면 **Pull requests** 탭에서 **New pull request**를 누릅니다.

### 8-2 어디에서 어디로 보낼지 고르기

화면 위쪽에 회색 상자 네 개가 한 줄로 있습니다. 화살표 `←` 왼쪽이 받는 쪽, 오른쪽이 보내는 쪽입니다.

| 상자 | 고를 값 |
|---|---|
| base repository | `Drillcheck/git-practice` |
| base | `mission/m01` |
| head repository | `<내 아이디>/git-practice` |
| compare | `m01-typo` |

**base는 처음에 `main`으로 되어 있습니다. 반드시 `mission/m01`로 바꿉니다.** 바꾸면 아래에 바뀐 파일이 `README.md` 하나, 한 줄만 보여야 합니다.

### 8-3 제목과 설명 쓰고 만들기

1. 제목은 커밋 메시지가 자동으로 들어갑니다. 그대로 둡니다.
2. 설명 칸에는 무엇을 왜 바꿨는지 한두 문장 씁니다. 예: "README 마지막 줄의 recieve를 receive로 고쳤습니다."
3. **Create pull request**를 누릅니다.

### 8-4 check 결과 보기

PR 화면 아래쪽에 check가 돌기 시작합니다. 1분쯤 지나면 결과가 나옵니다.

- 초록색 체크 표시와 **All checks have passed**: 통과입니다. m01 완료입니다.
- 빨간색 X: **Details**를 누르면 5-1에서 본 것과 같은 채점 결과가 나옵니다. 실패한 줄을 보고 고친 뒤, 커밋하고 `git push`하면 같은 PR에서 다시 채점됩니다.

### 8-5 알아 둘 것

- **Pull Request는 공개됩니다.** 내 GitHub 아이디와 커밋 내용이 누구에게나 보입니다. 원하지 않으면 PR을 열지 않고 `./check`로 내 컴퓨터에서만 채점해도 됩니다.
- 이 PR은 merge되지 않습니다. 다른 사람도 같은 미션을 같은 시작 상태에서 풀어야 하기 때문입니다. 실제 팀에서는 리뷰를 받고 merge되는 것으로 끝납니다.

## 10. Signed-off-by: "이 코드를 보낼 권리가 있습니다"

리눅스 커널 같은 프로젝트는 커밋마다 맨 끝에 이런 줄을 요구합니다.

```
Signed-off-by: Kim Minji <minji@example.invalid>
```

이 줄은 **DCO**(Developer Certificate of Origin)라는 짧은 약속에 동의한다는 표시입니다. "이 코드는 내가 썼거나, 이 프로젝트의 라이선스로 내놓을 권리가 있는 코드다"라는 뜻입니다.
이런 프로젝트는 PR에 DCO 검사를 걸어 두어서, 이 줄이 없는 커밋이 하나라도 있으면 check가 실패합니다.

### 10-1 앞으로 만들 커밋에 달기: git commit -s

새 커밋을 만들 때 `-s`를 붙이면 git이 이 줄을 붙여 줍니다. 이름과 메일은 00장에서 설정한 `user.name`, `user.email`에서 가져옵니다.

```
git commit -s -m "Fix ..."
```

### 10-2 이미 만든 커밋에 달기

방금 만든 두 커밋에는 이 줄이 없습니다. 마지막 커밋 하나라면 03장에서 배울 `--amend`로 고칠 수 있지만, 이번에는 두 개입니다.
`git rebase --signoff HEAD~2`는 "최근 커밋 2개를 다시 만들면서 각각에 이 줄을 붙여라"라는 뜻입니다. rebase는 04장과 05장에서 자세히 배웁니다.

입력:

```
git rebase --signoff HEAD~2
git log -2
```

결과:

```
Current branch fix-unknown-dish is up to date, rebase forced.
Successfully rebased and updated refs/heads/fix-unknown-dish.
commit ????????????????????????????????????????
Author: Kim Minji <minji@example.invalid>
Date:   Fri Nov 20 10:00:00 2026 +0900

    Sort the menu by dish name
    
    Signed-off-by: Kim Minji <minji@example.invalid>

commit ????????????????????????????????????????
Author: Kim Minji <minji@example.invalid>
Date:   Fri Nov 20 10:00:00 2026 +0900

    Reject dishes that are not on the menu
    
    A dish name that is not in MENU, such as a typo at the counter, stopped the calculator with a KeyError. Check every name first and say which one is wrong.
    
    Fixes #12
    
    Reported-by: Lee Jun <jun@example.invalid>
    Signed-off-by: Kim Minji <minji@example.invalid>
```

첫 줄은 "옮길 곳은 이미 그대로지만, `--signoff` 때문에 커밋을 다시 만들었다"는 뜻입니다. 두 커밋 모두 맨 끝에 `Signed-off-by:` 줄이 생겼습니다. 메시지가 바뀌었으니 커밋 ID도 새로 바뀌었습니다.

### 10-3 sign-off와 서명은 다릅니다

이름이 비슷해서 자주 헷갈립니다.

| | sign-off (`-s`, 소문자) | 서명 (`-S`, 대문자) |
|---|---|---|
| 무엇인가 | 메시지 끝에 붙는 글 한 줄 | 내 열쇠(GPG 또는 SSH 키)로 커밋에 붙이는 암호 서명 |
| 증명하는 것 | "보낼 권리가 있다"는 약속 | "이 커밋은 정말 이 사람이 만들었다" |
| GitHub에서 보이는 것 | 메시지 안의 `Signed-off-by:` 줄 | 커밋 옆의 **Verified** 표시 |
| 준비 | 없음 | 키를 만들고 GitHub에 등록. git 2.34부터는 SSH 키로도 서명할 수 있습니다 |

이 과정에서는 서명을 쓰지 않습니다. 회사나 프로젝트가 요구하면 그때 설정합니다.

## 11. 알아 두면 편한 요즘 기능

### 11-1 처음 push할 때도 git push만

7-1에서는 처음 push할 때 `-u origin 이름`을 붙였습니다. 붙이지 않으면 어떻게 되는지 봅니다.

입력:

```
git push
```

결과:

```
fatal: The current branch fix-unknown-dish has no upstream branch.
To push the current branch and set the remote as upstream, use

    git push --set-upstream origin fix-unknown-dish

To have this happen automatically for branches without a tracking
upstream, see 'push.autoSetupRemote' in 'git help config'.
```

내 브랜치가 fork의 어느 브랜치와 짝인지 아직 정하지 않았다는 오류입니다. 오류 안에 7-1과 같은 명령이 나옵니다.

git 2.37부터는 설정 하나로 이 짝짓기를 자동으로 하게 할 수 있습니다. "처음 push하는 브랜치는 fork에 같은 이름으로 올리고 짝지어라"라는 뜻입니다.

입력:

```
git config --global push.autoSetupRemote true
git push
```

결과:

```
To https://github.com/<내 아이디>/git-practice.git
 * [new branch]      fix-unknown-dish -> fix-unknown-dish
branch 'fix-unknown-dish' set up to track 'origin/fix-unknown-dish'.
```

마지막 줄의 `set up to track`이 7-1의 `-u`가 한 것과 같은 일입니다. 이 설정은 켜 두어도 되고, 켜지 않고 `-u`를 계속 써도 됩니다. 이 과정의 설명은 `-u`를 쓰는 방법으로 계속합니다.

### 11-2 터미널에서 PR 열기: gh pr create

8절에서는 브라우저에서 PR을 열었습니다. 00장에서 설치하고 로그인한 `gh`로는 터미널에서도 열 수 있습니다. 예를 들어 m01 PR이라면 이렇게 씁니다.

```
gh pr create --repo Drillcheck/git-practice --base mission/m01 --head <내 아이디>:m01-typo --fill
```

| 부분 | 뜻 |
|---|---|
| `--repo`, `--base` | 받는 쪽. 8-2의 base repository와 base |
| `--head` | 보내는 쪽. `아이디:브랜치` 모양으로 씁니다 |
| `--fill` | 제목과 설명을 커밋 메시지로 채웁니다 |

`gh auth login`으로 로그인해 둔 상태여야 합니다. 이번 연습 브랜치 `fix-unknown-dish`로는 PR을 열지 않습니다. 처음에는 8절처럼 브라우저에서 base를 눈으로 확인하는 쪽이 안전합니다.

### 11-3 연습용 브랜치 정리하기

연습용 브랜치를 내 컴퓨터와 fork에서 모두 지웁니다. fork의 브랜치를 지우는 명령은 05장에서 다시 다룹니다.

입력:

```
git switch m01-typo
git branch -D fix-unknown-dish
git push origin --delete fix-unknown-dish
```

결과:

```
Switched to branch 'm01-typo'
Your branch is up to date with 'origin/m01-typo'.
Deleted branch fix-unknown-dish (was ???????).
To https://github.com/<내 아이디>/git-practice.git
 - [deleted]         fix-unknown-dish
```

## 정리

| 하고 싶은 것 | 명령 |
|---|---|
| 새 브랜치를 만들고 이동 | `git switch -c 이름` |
| 다른 브랜치에서 갈라져 나오기 | `git switch -c 이름 시작브랜치` |
| 지금 브랜치 보기 | `git branch` |
| 바뀐 파일 목록 | `git status` |
| 아직 고르지 않은 변경 보기 | `git diff` |
| 다음 커밋에 넣을 변경 고르기 | `git add 파일` |
| 고른 변경 보기 | `git diff --staged` |
| 커밋 | `git commit -m "Fix ..."` |
| 아직 push하지 않은 브랜치를 처음부터 다시 | `git switch 시작브랜치`, 그다음 `git switch -C 이름` |
| 브랜치 지우기 | `git branch -D 이름` |
| 처음 push하면서 짝짓기 | `git push -u origin 이름` |
| 그다음부터 push | `git push` |
| 처음 push도 `git push`만으로 | `git config --global push.autoSetupRemote true` |
| 미션 채점 | `./check` |
| 파일 몇 개만 골라 커밋 | `git add 파일1 파일2`, 그다음 `git commit` |
| 제목, 본문, 꼬리말이 있는 메시지 | `git commit -m "제목" -m "본문" -m "Fixes #12"` 또는 `git commit`(편집기) |
| 메시지 전체 보기 | `git log -1` |
| Signed-off-by 붙여 커밋 | `git commit -s` |
| 최근 커밋 n개에 Signed-off-by 붙이기 | `git rebase --signoff HEAD~n` |
| 터미널에서 PR 열기 | `gh pr create --repo ... --base ... --head ...` |
